```python
# -*- coding: utf-8 -*-
"""
华泰期货《国债期货席位择时能力的动态筛选》(2025-07-29) 复现代码
运行环境: 聚宽研究环境 (Python 3.6.7 / numpy 1.19.5 / pandas 0.23.4 / scipy 1.2.0)
说明:
  1. 数据全部来自聚宽: 龙虎榜 finance.FUT_MEMBER_POSITION_RANK、get_dominant_future、get_price
  2. 不写磁盘: 数据在内存中直接使用, 结果直接 print (图用 plt.show 内联显示, 不保存文件)
  3. 写法兼容旧版本: 无 f-string / to_numpy / dataclass / 海象运算符等
  4. 直接把整段代码粘贴到研究环境的一个 cell 中运行即可

方法复现要点 (与研报对应):
  - 蜘蛛网席位信号: 每个席位把品种下所有合约的 "持买单量增减" 加总为 dB, "持卖单量增减" 加总为 dS
        dB>0 且 dS<0 -> 该席位发出多头信号 ; dB<0 且 dS>0 -> 该席位发出空头信号 ; 否则无信号
  - 多空方向分开回测每个席位:  只用多头信号的夏普 shL , 只用空头信号的夏普 shS
        多头优秀席位 = shL 前 N      -> 取其多头信号
        空头优秀席位 = shS 前 N      -> 取其空头信号
        多头反指席位 = shS 后 N      -> 其空头信号取反 => 做多   (研报: 做其对手方)
        空头反指席位 = shL 后 N      -> 其多头信号取反 => 做空
  - 合成信号 (等权):
        优秀信号: 只多/多空 ; 反指信号: 只多/多空 ; 复合信号: 只多/多空 (= 优秀与反指等权合成)
  - 滚动筛选: 每 X 天, 用过去 Y 天数据重新筛选, 选前N/后N, 在接下来 X 天内使用
        X in (5,10,20,60), Y in (120,240,480), N in (1,3,5,10)  (TL: Y in (120,240))
  - 复利, 不计滑点和手续费; 信号收盘后产生, 下一交易日持有 (无未来函数)
  - 与 "固定样本期" 方法对比: 固定样本期=回测开始之前的数据 (一次性筛选, N=FIXED_N)

注意 (与研报不同之处, 因为数据区间不同):
  - 回测期 2024-01-01 ~ 2026-12-20 (超过今天的部分自动截断)
  - 滚动筛选需要回看 Y(最长480)天历史, 所以数据从回测开始前 WARMUP_DAYS 天开始取 (仅作预热, 不计入回测收益)
  - 收益使用 "前一日收盘时的主力合约" 的收盘价涨跌幅, 避免主力换月跳空
  - 席位当天不在前20名时, 其增减量按 0 处理
  - 参数组合/信号类型的 "最优" 是按回测期夏普在样本内选出的, 带有过拟合, 仅为复现研报做法
"""
import re
import time
import datetime
import numpy as np
import pandas as pd

from jqdata import *
from jqdata import finance

# ========================= 参数区 =========================
PRODUCTS = [('TS', '两年期'), ('TF', '五年期'), ('T', '十年期'), ('TL', '三十年期')]
BT_START = datetime.date(2024, 1, 1)      # 回测开始
BT_END = datetime.date(2026, 12, 20)      # 回测结束 (超过今天自动截断)
WARMUP_DAYS = 760                         # 预热天数(自然日), 480个交易日约 700 自然日
X_LIST = (5, 10, 20, 60)                  # 换仓频率
Y_LIST = (120, 240, 480)                  # 回溯周期
TL_Y_LIST = (120, 240)                    # 30年期上市晚, 只测 120/240
N_LIST = (1, 3, 5, 10)                    # 排名选择
FIXED_N = 5                               # 固定样本期方法的前N/后N
ANN = 252                                 # 年化天数
MIN_SIGS = 5                              # 窗口内席位至少有多少个信号才参与排名
MIN_WIN = 60                              # 窗口至少多少个交易日才筛选
SELECT_BY = 'sharpe'                      # 选最优组合的指标: 'sharpe' 或 'calmar'
PLOT = True                               # 是否内联画净值图(不保存文件)
CHUNK_DAYS = 30                           # 龙虎榜按多少自然日分段取(单次最多返回4000行, 超了会自动二分)

SIG_NAMES = ['优秀信号:只多', '优秀信号:多空', '反指信号:只多',
             '反指信号:多空', '复合信号:只多', '复合信号:多空']
LONG_IDX = [0, 2, 4]
LS_IDX = [1, 3, 5]

pd.set_option('display.width', 220)
pd.set_option('display.max_columns', 30)
pd.set_option('display.unicode.east_asian_width', True)


# ========================= 数据获取 =========================
def _run(u, d0, d1):
    """取一段日期内某品种所有合约的 持买单量/持卖单量 排名 (rank_type_ID 501002/501003)"""
    T = finance.FUT_MEMBER_POSITION_RANK
    q = query(T.day, T.code, T.rank_type_ID, T.rank, T.member_name, T.indicator_increase
              ).filter(T.code.like(u + '2%'),
                       T.rank_type_ID.in_([501002, 501003]),
                       T.day >= d0, T.day <= d1)
    return finance.run_query(q)


def _fetch(u, d0, d1, out):
    df = _run(u, d0, d1)
    if len(df) >= 4000 and d1 > d0:      # 触及4000行上限, 二分重取
        mid = d0 + datetime.timedelta(days=(d1 - d0).days // 2)
        _fetch(u, d0, mid, out)
        _fetch(u, mid + datetime.timedelta(days=1), d1, out)
    elif len(df) > 0:
        out.append(df)


def load_rank(u, d0, d1):
    out = []
    cur = d0
    while cur <= d1:
        nxt = min(cur + datetime.timedelta(days=CHUNK_DAYS - 1), d1)
        _fetch(u, cur, nxt, out)
        cur = nxt + datetime.timedelta(days=1)
    if len(out) == 0:
        return None
    df = pd.concat(out, ignore_index=True)
    pat = re.compile('^' + u + r'\d{4}\.CCFX$')      # 只保留该品种(排除 TS/TF/TL 与 T 混淆)
    df = df[df['code'].apply(lambda c: pat.match(str(c)) is not None)].copy()
    df['day'] = pd.to_datetime(df['day'])
    df['indicator_increase'] = pd.to_numeric(df['indicator_increase'], errors='coerce').fillna(0.0)
    return df


def build_seat_matrices(df):
    """-> dates(DatetimeIndex), seats(Index), dB矩阵, dS矩阵 (日期 x 席位)"""
    b = df[df['rank_type_ID'] == 501002].groupby(['day', 'member_name'])['indicator_increase'].sum().unstack('member_name')
    s = df[df['rank_type_ID'] == 501003].groupby(['day', 'member_name'])['indicator_increase'].sum().unstack('member_name')
    dates = b.index.union(s.index)
    seats = b.columns.union(s.columns)
    b = b.reindex(index=dates, columns=seats).fillna(0.0)
    s = s.reindex(index=dates, columns=seats).fillna(0.0)
    return dates, seats, b.values, s.values


def get_returns(u, dates):
    """
    ret[k] = 第k个日期当天实现的收益: 持有 "前一日(收盘时)主力合约" 的收盘价涨跌幅
    """
    n = len(dates)
    ret = np.zeros(n)
    try:
        dom = []
        for d in dates:
            dom.append(get_dominant_future(u, d.strftime('%Y-%m-%d')))
        closes = {}
        for c in set(dom):
            if c and c not in closes:
                p = get_price(c, start_date=dates[0].strftime('%Y-%m-%d'),
                              end_date=dates[-1].strftime('%Y-%m-%d'),
                              frequency='daily', fields=['close'], skip_paused=True)
                closes[c] = p['close']
        for k in range(1, n):
            c = dom[k - 1]
            if (not c) or (c not in closes):
                continue
            px = closes[c]
            p0 = px.get(dates[k - 1], np.nan)
            p1 = px.get(dates[k], np.nan)
            if p0 == p0 and p1 == p1 and p0 > 0:
                ret[k] = p1 / p0 - 1.0
    except Exception as e:
        print('  [警告] 主力合约逐日取价失败(%s), 退回使用 %s9999.CCFX 连续合约涨跌幅(含换月跳空)' % (str(e), u))
        p = get_price(u + '9999.CCFX', start_date=dates[0].strftime('%Y-%m-%d'),
                      end_date=dates[-1].strftime('%Y-%m-%d'), frequency='daily',
                      fields=['close'], skip_paused=True)['close']
        ret = p.pct_change().reindex(dates).fillna(0.0).values
    return ret


# ========================= 滚动筛选引擎 =========================
def prep(L, Sg, R):
    """前缀和, 使任意窗口内每个席位的 夏普 可以 O(1) 算出"""
    S = L.shape[1]
    z = np.zeros((1, S))
    SL = L * R[:, None]          # 席位"只用多头信号"的日收益
    SS = -Sg * R[:, None]        # 席位"只用空头信号"的日收益

    def pre(a):
        return np.vstack([z, np.cumsum(a, axis=0)])
    return {'L': L, 'S': Sg, 'R': R,
            'l1': pre(SL), 'l2': pre(SL ** 2), 'lc': pre(L),
            's1': pre(SS), 's2': pre(SS ** 2), 'sc': pre(Sg)}


def _sharpe(p1, p2, pc, a, b):
    m = float(b - a)
    mean = (p1[b] - p1[a]) / m
    var = (p2[b] - p2[a]) / m - mean ** 2
    cnt = pc[b] - pc[a]
    sh = np.full(mean.shape, np.nan)
    ok = (var > 1e-12) & (cnt >= MIN_SIGS)
    sh[ok] = mean[ok] / np.sqrt(var[ok]) * np.sqrt(ANN)
    return sh


def _pick(sh, N):
    """返回 (前N等权权重向量, 后N等权权重向量)"""
    S = len(sh)
    wt = np.zeros(S)
    wb = np.zeros(S)
    idx = np.where(~np.isnan(sh))[0]
    if len(idx) < 2:
        return wt, wb
    k = min(N, len(idx) // 2)
    order = idx[np.argsort(sh[idx])]
    wb[order[:k]] = 1.0 / k
    wt[order[-k:]] = 1.0 / k
    return wt, wb


def rolling_positions(D, i0, X, Y, n_list):
    """
    从 i0 起每 X 天换仓, 用 [i-Y, i) 的数据筛选, 作用于信号日 [i, i+X)
    返回 {N: 6 x n 的仓位矩阵}, 行顺序同 SIG_NAMES
    """
    n = D['L'].shape[0]
    out = {}
    for N in n_list:
        out[N] = np.zeros((6, n))
    i = i0
    while i < n:
        j = min(i + X, n)
        a = max(0, i - Y)
        b = i
        if b - a >= MIN_WIN:
            shL = _sharpe(D['l1'], D['l2'], D['lc'], a, b)
            shS = _sharpe(D['s1'], D['s2'], D['sc'], a, b)
            Lb = D['L'][i:j]
            Sb = D['S'][i:j]
            for N in n_list:
                topL, botL = _pick(shL, N)
                topS, botS = _pick(shS, N)
                A_long = Lb.dot(topL)         # 多头优秀席位的多头信号
                A_short = -Sb.dot(topS)       # 空头优秀席位的空头信号
                B_long = Sb.dot(botS)         # 多头反指席位: 空头信号取反 -> 多
                B_short = -Lb.dot(botL)       # 空头反指席位: 多头信号取反 -> 空
                P = out[N]
                P[0, i:j] = A_long
                P[1, i:j] = A_long + A_short
                P[2, i:j] = B_long
                P[3, i:j] = B_long + B_short
                P[4, i:j] = (A_long + B_long) / 2.0
                P[5, i:j] = (A_long + A_short + B_long + B_short) / 2.0
        i += X
    return out


# ========================= 绩效 =========================
def fast_perf(r):
    n = len(r)
    if n < 20:
        return None
    sd = r.std(ddof=1)
    if sd < 1e-12:
        return None
    nav = np.cumprod(1.0 + r)
    cum = nav[-1] - 1.0
    ann = nav[-1] ** (ANN / float(n)) - 1.0
    sharpe = r.mean() / sd * np.sqrt(ANN)
    peak = np.maximum.accumulate(np.concatenate([[1.0], nav]))[1:]
    mdd = (nav / peak - 1.0).min()
    calmar = ann / abs(mdd) if mdd < 0 else np.nan
    return {'cum': cum, 'ann': ann, 'sharpe': sharpe, 'calmar': calmar, 'mdd': mdd}


def full_perf(r):
    """r: pd.Series (日收益, index为日期)"""
    n = len(r)
    nav = (1.0 + r).cumprod()
    cum = nav.iloc[-1] - 1.0
    ann = nav.iloc[-1] ** (ANN / float(n)) - 1.0
    sd = r.std()
    sharpe = r.mean() / sd * np.sqrt(ANN) if sd > 0 else np.nan
    peak = nav.cummax().clip(lower=1.0)
    dd = nav / peak - 1.0
    mdd = dd.min()
    end = dd.idxmin()
    start = nav[:end].idxmax()
    calmar = ann / abs(mdd) if mdd < 0 else np.nan
    return [cum, ann, sharpe, calmar, mdd, start, end]


def perf_table(rows):
    """rows: [(label, pd.Series)]"""
    cols = ['累计收益率', '年化收益率', '夏普比率', '卡玛比率', '最大回撤', '回撤开始时间', '回撤结束时间']
    data = []
    idx = []
    for label, s in rows:
        p = full_perf(s)
        data.append(['%.2f%%' % (p[0] * 100), '%.2f%%' % (p[1] * 100), '%.4f' % p[2], '%.4f' % p[3],
                     '%.2f%%' % (p[4] * 100), str(p[5].date()), str(p[6].date())])
        idx.append(label)
    return pd.DataFrame(data, index=idx, columns=cols)


def plot_nav(title, series_dict):
    try:
        import matplotlib.pyplot as plt
        plt.figure(figsize=(11, 4))
        for k in series_dict:
            s = series_dict[k]
            (1.0 + s).cumprod().plot(label=k)
        plt.title(title)
        plt.legend(loc='best', fontsize=8)
        plt.grid(True, alpha=0.3)
        plt.show()
    except Exception as e:
        print('  [画图失败] ' + str(e))


# ========================= 主流程 =========================
def run_product(u, name, d0, d1):
    print('=' * 100)
    print('【%s国债期货 %s】 数据区间 %s ~ %s (含预热), 回测起点 %s' % (name, u, d0, d1, BT_START))
    t1 = time.time()
    df = load_rank(u, d0, d1)
    if df is None or len(df) == 0:
        print('  没有取到龙虎榜数据, 跳过')
        return None
    dates, seats, B, Sd = build_seat_matrices(df)
    n = len(dates)
    print('  龙虎榜: %d 个交易日, %d 个席位, 用时 %.1fs' % (n, len(seats), time.time() - t1))

    L = ((B > 0) & (Sd < 0)).astype(float)       # 席位多头信号
    Sg = ((B < 0) & (Sd > 0)).astype(float)      # 席位空头信号
    ret = get_returns(u, dates)
    R = np.append(ret[1:], 0.0)                  # R[t]: 信号日t之后下一交易日的收益
    i0 = int(dates.searchsorted(pd.Timestamp(BT_START)))
    if i0 >= n - 30:
        print('  回测期内数据不足, 跳过')
        return None
    D = prep(L, Sg, R)

    rng = slice(i0, n - 1)
    ret_idx = dates[i0 + 1:n]
    hold = pd.Series(R[rng], index=ret_idx)
    ys = TL_Y_LIST if u == 'TL' else Y_LIST

    # ---- 参数网格 ----
    results = []
    for X in X_LIST:
        for Y in ys:
            posd = rolling_positions(D, i0, X, Y, N_LIST)
            for N in N_LIST:
                P = posd[N]
                for k in range(6):
                    m = fast_perf(P[k, rng] * R[rng])
                    if m is not None:
                        results.append((k, X, Y, N, m))
    print('  参数网格测试完成: %d 个组合有效, 累计用时 %.1fs' % (len(results), time.time() - t1))

    # ---- 固定样本期(回测开始前一次性筛选) ----
    Pfix = rolling_positions(D, i0, n, i0, [FIXED_N])[FIXED_N]

    summary = []
    for mode_name, idxs in [('多头择时(只多头)', LONG_IDX), ('多空择时', LS_IDX)]:
        cand = [r for r in results if r[0] in idxs]
        if len(cand) == 0:
            continue
        key = 'sharpe' if SELECT_BY == 'sharpe' else 'calmar'
        vals = np.array([c[4][key] for c in cand], dtype=float)
        if np.all(np.isnan(vals)):
            continue
        order = np.argsort(-np.where(np.isnan(vals), -1e9, vals))
        best = cand[order[0]]
        k, X, Y, N, m = best
        print('-' * 100)
        print('>> %s: 最优 = %s, 参数 (X=%d, Y=%d, N=%d)   [按回测期%s选取]' % (mode_name, SIG_NAMES[k], X, Y, N, SELECT_BY))
        print('   网格中该模式 %s 前5名:' % SELECT_BY)
        for q in order[:5]:
            c = cand[q]
            print('     %-10s X=%-3d Y=%-3d N=%-2d  夏普=%.4f  卡玛=%.4f  累计=%.2f%%' %
                  (SIG_NAMES[c[0]], c[1], c[2], c[3], c[4]['sharpe'], c[4]['calmar'], c[4]['cum'] * 100))

        Pbest = rolling_positions(D, i0, X, Y, [N])[N]
        roll = pd.Series(Pbest[k, rng] * R[rng], index=ret_idx)
        rows = [('多头持有', hold)]
        plot_d = {'Hold': hold}
        for kk, tag in zip(idxs, ['优秀', '反指', '复合']):
            s = pd.Series(Pfix[kk, rng] * R[rng], index=ret_idx)
            if s.std() > 0:
                rows.append(('固定-' + SIG_NAMES[kk], s))
                plot_d['Fixed ' + str(kk)] = s
        rows.append(('滚动-' + SIG_NAMES[k], roll))
        plot_d['Rolling best'] = roll
        print(perf_table(rows))
        if PLOT:
            plot_nav('%s %s (BT %s ~ %s)' % (u, 'long-only' if idxs is LONG_IDX else 'long-short',
                                              str(ret_idx[0].date()), str(ret_idx[-1].date())), plot_d)
        ph = full_perf(hold)
        pr = full_perf(roll)
        summary.append([u, mode_name, SIG_NAMES[k], 'X=%d,Y=%d,N=%d' % (X, Y, N),
                        '%.4f' % ph[2], '%.4f' % pr[2], '%.4f' % ph[3], '%.4f' % pr[3],
                        '%.2f%%' % (ph[1] * 100), '%.2f%%' % (pr[1] * 100)])
    return summary


def main():
    t0 = time.time()
    end = min(BT_END, datetime.date.today())
    d0 = BT_START - datetime.timedelta(days=WARMUP_DAYS)
    all_sum = []
    for u, name in PRODUCTS:
        try:
            s = run_product(u, name, d0, end)
            if s:
                all_sum.extend(s)
        except Exception as e:
            import traceback
            print('  [%s 运行出错] %s' % (u, str(e)))
            traceback.print_exc()
    print('=' * 100)
    print('汇总: 滚动合成信号 vs 多头持有 (回测期 %s ~ %s)' % (BT_START, end))
    if all_sum:
        cols = ['品种', '模式', '最优信号', '参数', '持有夏普', '滚动夏普', '持有卡玛', '滚动卡玛', '持有年化', '滚动年化']
        print(pd.DataFrame(all_sum, columns=cols))
    print('总用时 %.1fs' % (time.time() - t0))


if __name__ == '__main__':
    main()
```
