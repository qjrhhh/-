```python
# -*- coding: utf-8 -*-
"""
复现：华泰期货《风格轮动下的 A500 对冲策略》(2025-10-31)
运行环境：聚宽研究环境（Python 3.6 / pandas 0.23 / numpy 1.19 / scipy 1.2）
区间：2024-01-01 ~ 2026-12-20（超出数据末端自动截断）
说明：数据直接取用、结果直接 print / 画图，不写磁盘。
"""
import datetime
import warnings
import numpy as np
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt

warnings.filterwarnings('ignore')
try:
    from jqdata import *
except Exception:
    pass

pd.set_option('display.width', 220)
pd.set_option('display.max_columns', 30)

# ============================ 参数 ============================
START = '2024-01-01'                       # 评价起点
END_WANT = datetime.date(2026, 12, 20)     # 评价终点
END = min(END_WANT, datetime.date.today()).strftime('%Y-%m-%d')
WARM_START = '2023-01-01'                  # 预热期(滚动窗口需要历史数据)
WIN = 120                                  # "滚动半年" ≈ 120 个交易日
CORR_TH = 0.85                             # 相关性因子阈值
MIN_DAYS = 10                              # 合约轮动：距到期至少 N 个自然日
PLOT = True

A500_CODE = '000510.XSHG'                  # 中证A500
SPOT = {'IH': '000016.XSHG', 'IF': '000300.XSHG',
        'IC': '000905.XSHG', 'IM': '000852.XSHG'}
PRODS = ['IH', 'IF', 'IC', 'IM']
HEDGE_PRODS = ['IH', 'IF', 'IC']


def pct(x):
    return '%.2f%%' % (x * 100.0)


def get_close(code, s, e):
    df = get_price(code, start_date=s, end_date=e, frequency='daily',
                   fields=['close'], skip_paused=False, fq='pre')
    return df['close']


# ============================ 1. 基础数据 ============================
a500 = get_close(A500_CODE, WARM_START, END).dropna()
if len(a500) == 0:
    raise Exception('取不到中证A500(000510.XSHG)行情，请检查数据权限')
dates = a500.index
print('A500 行情区间: %s ~ %s, 共 %d 个交易日' % (dates[0].date(), dates[-1].date(), len(dates)))
if dates[0] > pd.Timestamp(WARM_START) + pd.Timedelta(days=10):
    print('提示: A500 指数历史数据起点晚于预热起点，滚动因子的有效起点会相应推后')

spots = {}
for p, c in SPOT.items():
    spots[p] = get_close(c, WARM_START, END).reindex(dates)
spot_ret = pd.DataFrame(spots).pct_change()
a500_ret = a500.pct_change()

# ============================ 2. 合约信息 ============================
allf = get_all_securities(['futures'])
tdays = pd.DatetimeIndex(get_all_trade_days())


def third_friday_expiry(year, mon):
    d0 = pd.Timestamp(year, mon, 1)
    fr = []
    for i in range(31):
        d = d0 + pd.Timedelta(days=i)
        if d.month == mon and d.weekday() == 4:
            fr.append(d)
    tf = fr[2]
    pos = tdays.searchsorted(tf)
    if pos < len(tdays):
        return tdays[pos]
    return tf


rows = []
for code in allf.index:
    if (not code.endswith('.CCFX')) or len(code) != 11:
        continue
    prod = code[:2]
    if prod not in PRODS:
        continue
    ym = code[2:6]
    if not ym.isdigit():
        continue
    mon = int(ym[2:])
    if mon < 1 or mon > 12:
        continue
    s = pd.Timestamp(allf.loc[code, 'start_date'])
    exp = third_friday_expiry(2000 + int(ym[:2]), mon)   # 到期日=第三个周五(遇假日顺延)
    if exp < pd.Timestamp(WARM_START) or s > pd.Timestamp(END):
        continue
    rows.append((code, prod, mon, s, exp, mon in (3, 6, 9, 12)))
info = pd.DataFrame(rows, columns=['code', 'prod', 'mon', 'start', 'exp', 'quarter']).set_index('code')

need = info[(info['prod'].isin(HEDGE_PRODS)) | (info['quarter'])]
px = {}
for code in need.index:
    s = max(need.loc[code, 'start'], dates[0])
    e = min(need.loc[code, 'exp'], dates[-1])
    if s > e:
        continue
    try:
        px[code] = get_close(code, s.strftime('%Y-%m-%d'), e.strftime('%Y-%m-%d'))
    except Exception as ex:
        print('获取合约失败', code, ex)
P = pd.DataFrame(px).reindex(dates)
Pv = P.values
colpos = dict((c, i) for i, c in enumerate(P.columns))
print('共获取 %d 个期货合约的日线数据' % P.shape[1])

# ---------- 当季合约连续收益 ----------
def quarter_return(prod):
    qs = info[(info['prod'] == prod) & (info['quarter'])].sort_values('exp')
    qs = [(c, r['start'], r['exp']) for c, r in qs.iterrows() if c in colpos]
    n = len(dates)
    ret = np.full(n, np.nan)
    pick = [None] * n
    for i in range(1, n):
        t, tp = dates[i], dates[i - 1]
        for c, s, e in qs:
            if e >= t and s <= tp:
                j = colpos[c]
                pick[i] = c
                if not (np.isnan(Pv[i, j]) or np.isnan(Pv[i - 1, j])):
                    ret[i] = Pv[i, j] / Pv[i - 1, j] - 1.0
                break
    return pd.Series(ret, index=dates), pd.Series(pick, index=dates)


Q = {}
for p in PRODS:
    Q[p], _ = quarter_return(p)
    miss = Q[p].iloc[1:].isnull().sum()
    if miss > 0:
        print('警告: %s 当季合约有 %d 天缺少收益，已按0处理' % (p, miss))
QR = pd.DataFrame(Q).fillna(0.0)                 # 期货(当季)日收益
R = pd.concat([a500_ret.rename('A500'), QR], axis=1).iloc[1:]


# ---------- 基差轮动：在 4 个可交易合约里选年化基差最高(空头最划算)的 ----------
def rotate_return(prod):
    cands = [(c, r['start'], r['exp']) for c, r in info[info['prod'] == prod].iterrows() if c in colpos]
    spot = spots[prod].values
    n = len(dates)
    ret = np.full(n, np.nan)
    for i in range(1, n):
        t, tp = dates[i], dates[i - 1]
        best, bv = None, -1e18
        for c, s, e in cands:
            if s <= tp and e >= t:
                j = colpos[c]
                if np.isnan(Pv[i - 1, j]) or np.isnan(Pv[i, j]) or np.isnan(spot[i - 1]):
                    continue
                days = (e - tp).days
                if days < MIN_DAYS:
                    continue
                ab = (Pv[i - 1, j] / spot[i - 1] - 1.0) * 365.0 / days
                if ab > bv:
                    bv, best = ab, j
        if best is not None:
            ret[i] = Pv[i, best] / Pv[i - 1, best] - 1.0
    s = pd.Series(ret, index=dates)
    return s.fillna(QR[prod])                     # 无候选时退回当季合约


ROT = pd.DataFrame(dict((p, rotate_return(p)) for p in HEDGE_PRODS))

# ============================ 3. 因子与信号 ============================
roll_corr = dict((p, R['A500'].rolling(WIN).corr(R[p])) for p in PRODS)
roll_vol = dict((p, R[p].rolling(WIN).std()) for p in ['A500'] + PRODS)

cb = pd.DataFrame({'IH': roll_corr['IH'], 'IC': roll_corr['IC']})
valid_c = cb.notnull().all(axis=1)
best_c = np.where(cb['IH'] >= cb['IC'], 'IH', 'IC')
sig_corr = pd.Series(np.where(cb.max(axis=1) >= CORR_TH, best_c, 'IF'), index=cb.index).where(valid_c)

dv = pd.DataFrame(dict((p, (roll_vol[p] - roll_vol['A500']).abs()) for p in HEDGE_PRODS))
valid_v = dv.notnull().all(axis=1)
sig_vol = dv.idxmin(axis=1).where(valid_v)

cp = np.where(cb['IH'] > cb['IC'], 'IH', 'IC')
vp = np.where(dv['IH'] < dv['IC'], 'IH', 'IC')
sig_mix = pd.Series(np.where(cp == vp, cp, 'IF'), index=cb.index).where(valid_c & valid_v)

# 信号 t-1 日收盘后产生，t 日生效
sig_corr, sig_vol, sig_mix = sig_corr.shift(1), sig_vol.shift(1), sig_mix.shift(1)

first_ok = max(sig_corr.first_valid_index(), sig_vol.first_valid_index(), sig_mix.first_valid_index())
E_start = max(pd.Timestamp(START), first_ok)
E = R.index[R.index >= E_start]
print('\n评价区间: %s ~ %s (%d 个交易日)' % (E[0].date(), E[-1].date(), len(E)))


def pick_ret(pick, RET):
    cols = list(RET.columns)
    idx = np.array([cols.index(p) if p in cols else -1 for p in pick.values])
    vals = RET.values
    out = np.where(idx >= 0, vals[np.arange(len(idx)), np.where(idx >= 0, idx, 0)], np.nan)
    return pd.Series(out, index=pick.index)


# ============================ 4. 绩效函数 ============================
def perf(r, base_vol=None, base_ret=None):
    r = r.dropna()
    n = len(r)
    nav = np.concatenate([[1.0], np.cumprod(1.0 + r.values)])
    ann_ret = nav[-1] ** (252.0 / n) - 1.0
    vol = r.std() * np.sqrt(252.0)
    mdd = (nav / np.maximum.accumulate(nav) - 1.0).min()
    out = {'年化收益': ann_ret, '年化波动率': vol, '夏普比率': ann_ret / vol if vol > 0 else np.nan,
           '最大回撤': mdd, '年化对冲成本': 0.0, '套保比率': 0.0}
    if base_vol is not None:
        out['年化对冲成本'] = ann_ret - base_ret
        out['套保比率'] = 1.0 - vol / base_vol
    return out


def make_table(strats, base):
    pb = perf(base)
    rows, idx = [], []
    for name, r in strats:
        pr = perf(r, pb['年化波动率'], pb['年化收益'])
        rows.append(pr)
        idx.append(name)
    rows.append(pb)
    idx.append('无对冲')
    df = pd.DataFrame(rows, index=idx)[['年化收益', '年化波动率', '夏普比率', '最大回撤', '年化对冲成本', '套保比率']]
    out = df.copy()
    for c in out.columns:
        if c == '夏普比率':
            out[c] = df[c].map(lambda x: '%.2f' % x)
        else:
            out[c] = df[c].map(pct)
    return out


def plot_nav(title, series, sig=None):
    if not PLOT:
        return
    fig, ax = plt.subplots(figsize=(10, 4))
    for name, r in series.items():
        nav = (1.0 + r).cumprod()
        ax.plot(nav.index, nav.values, label=name)
    if sig is not None:
        ax2 = ax.twinx()
        ax2.fill_between(sig.index, sig.values, 0, color='grey', alpha=0.2, step='post')
        ax2.set_ylim(-1.2, 1.3)
    ax.legend(loc='best')
    ax.set_title(title)
    plt.show()


Re = R.loc[E]
base = Re['A500']
sigmap = {'IC': 1.0, 'IF': -1.0, 'IH': 0.5}

# ============================ 表1: 成分股对比 ============================
print('\n' + '=' * 30 + ' 表1 A500 与宽基指数成分股对比 ' + '=' * 30)
try:
    last_td = d2 = dates[-1].strftime('%Y-%m-%d')
    w500 = get_index_weights(A500_CODE, date=last_td)['weight']
    rows, idx = [], []
    for nm, code in [('上证50', '000016.XSHG'), ('沪深300', '000300.XSHG'),
                     ('中证500', '000905.XSHG'), ('中证1000', '000852.XSHG')]:
        w = get_index_weights(code, date=last_td)['weight']
        common = w.index.intersection(w500.index)
        rows.append({'相同成分股数量': len(common),
                     '指数相同成分权重': '%.2f%%' % w[common].sum(),
                     'A500相同成分权重': '%.2f%%' % w500[common].sum()})
        idx.append(nm)
    print('权重日期:', last_td)
    print(pd.DataFrame(rows, index=idx)[['相同成分股数量', '指数相同成分权重', 'A500相同成分权重']])
except Exception as ex:
    print('成分权重获取失败:', ex)

# ============================ 表2: 相关性 ============================
print('\n' + '=' * 30 + ' 表2 A500 与宽基指数日收益相关性 ' + '=' * 30)
rows, idx = [], []
for p in PRODS:
    x = pd.concat([a500_ret.loc[E], spot_ret[p].loc[E]], axis=1).dropna()
    r_, pv = stats.pearsonr(x.iloc[:, 0].values, x.iloc[:, 1].values)
    rows.append({'相关系数': '%.3f' % r_, '相关系数p-value': '%.3f' % pv})
    idx.append(p)
print(pd.DataFrame(rows, index=idx)[['相关系数', '相关系数p-value']])

# ============================ 表3 / 图3: 单品种对冲 ============================
print('\n' + '=' * 30 + ' 表3 单品种对冲 A500 (当季合约, 等市值, 不计成本) ' + '=' * 30)
single = [(p, base - Re[p]) for p in PRODS]
t3 = make_table([(('%s(额外)' % p if p == 'IM' else p), r) for p, r in single], base)
print(t3)
plot_nav('Fig3 single-product hedge', dict([('A500', base)] + [(p, r) for p, r in single if p != 'IM']))

# ============================ 表4 / 表5 / 图4: 回归多品种 ============================
def ols(y, X):
    n = len(y)
    X1 = np.column_stack([np.ones(n), X])
    beta = np.linalg.lstsq(X1, y, rcond=None)[0]
    resid = y - X1.dot(beta)
    k = X1.shape[1]
    s2 = resid.dot(resid) / (n - k)
    se = np.sqrt(np.diag(s2 * np.linalg.inv(X1.T.dot(X1))))
    t = beta / se
    pv = 2 * (1 - stats.t.cdf(np.abs(t), n - k))
    q = stats.t.ppf(0.975, n - k)
    r2 = 1 - resid.dot(resid) / ((y - y.mean()) ** 2).sum()
    return beta, t, pv, beta - q * se, beta + q * se, r2


print('\n' + '=' * 30 + ' 表4 回归结果 (A500日收益 ~ 期货日收益) ' + '=' * 30)
rows, idx = [], []
weights = {}
for combo in [('IH', 'IC'), ('IF', 'IC')]:
    y = Re['A500'].values
    X = Re[list(combo)].values
    beta, t, pv, lo, hi, r2 = ols(y, X)
    for k, v in enumerate(combo):
        rows.append({'回归变量': v, '回归系数': '%.4f' % beta[k + 1], 't': '%.3f' % t[k + 1],
                     'P>|t|': '%.3f' % pv[k + 1], '[0.025': '%.3f' % lo[k + 1],
                     '0.975]': '%.3f' % hi[k + 1], 'R2': '%.3f' % r2})
        idx.append('+'.join(combo))
    w1 = round(beta[1] / (beta[1] + beta[2]), 1)           # 按回归系数占比取整到 10%
    weights[combo] = (w1, 1.0 - w1)
print(pd.DataFrame(rows, index=idx)[['回归变量', '回归系数', 't', 'P>|t|', '[0.025', '0.975]', 'R2']])
for combo, w in weights.items():
    print('%s 组合配比: %s %.0f%%, %s %.0f%%' % ('+'.join(combo), combo[0], w[0] * 100, combo[1], w[1] * 100))

reg_ret = {}
for combo, w in weights.items():
    reg_ret['+'.join(combo)] = base - (w[0] * Re[combo[0]] + w[1] * Re[combo[1]])

print('\n' + '=' * 30 + ' 表5 多品种组合对冲 A500 ' + '=' * 30)
print(make_table([(k, reg_ret[k]) for k in ['IH+IC', 'IF+IC']], base))
plot_nav('Fig4 multi-product combo hedge', {'A500': base, 'IH+IC': reg_ret['IH+IC'], 'IF+IC': reg_ret['IF+IC']})

# ============================ 图5 / 图6: 滚动相关性 & 波动率 ============================
print('\n' + '=' * 30 + ' 图5/6 滚动半年相关性与波动率 (区间统计) ' + '=' * 30)
rc = pd.DataFrame(dict((p, roll_corr[p]) for p in PRODS)).loc[E]
rv = pd.DataFrame(dict((p, roll_vol[p]) for p in ['A500'] + PRODS)).loc[E] * 100.0
print('滚动相关性 min / max / 最新:')
print(pd.DataFrame({'min': rc.min(), 'max': rc.max(), 'last': rc.iloc[-1]}).round(3))
print('滚动日波动率(%) min / max / 最新:')
print(pd.DataFrame({'min': rv.min(), 'max': rv.max(), 'last': rv.iloc[-1]}).round(3))
if PLOT:
    ax = rc.plot(figsize=(10, 3.5), title='Fig5 rolling %d-day corr with A500' % WIN)
    plt.show()
    ax = rv.plot(figsize=(10, 3.5), title='Fig6 rolling %d-day daily vol (%%)' % WIN)
    plt.show()

# ============================ 表6 / 图7 / 图8: 单因子轮动 ============================
ret_corr = (R['A500'] - pick_ret(sig_corr, R[HEDGE_PRODS])).loc[E]
ret_vol = (R['A500'] - pick_ret(sig_vol, R[HEDGE_PRODS])).loc[E]
print('\n' + '=' * 30 + ' 表6 单因子风格轮动 ' + '=' * 30)
print(make_table([('IH+IF+IC(相关性)', ret_corr), ('IH+IF+IC(波动率)', ret_vol)], base))
plot_nav('Fig7 corr rotation (signal: IC=1, IF=-1, IH=0.5)',
         {'A500': base, 'rotation': ret_corr, 'IH': single[0][1], 'IF': single[1][1], 'IC': single[2][1]},
         sig_corr.loc[E].map(sigmap))
plot_nav('Fig8 vol rotation', {'A500': base, 'rotation': ret_vol, 'IH': single[0][1], 'IF': single[1][1], 'IC': single[2][1]},
         sig_vol.loc[E].map(sigmap))

# ============================ 表7 / 图9 / 图10: 双因子轮动 ============================
ret_mix = (R['A500'] - pick_ret(sig_mix, R[HEDGE_PRODS])).loc[E]
sm = sig_mix.loc[E]
nsw = int((sm != sm.shift(1)).sum()) - 1
print('\n多因子轮动：区间内切换 %d 次，平均每 %.1f 个交易日切换一次' % (nsw, len(sm) / float(max(nsw, 1))))
print('各品种持有天数占比:')
print((sm.value_counts() / float(len(sm))).round(3))

print('\n' + '=' * 30 + ' 表7 风格轮动 A500 对冲策略效果对比 ' + '=' * 30)
print(make_table([('IH+IF+IC', ret_mix), ('IH+IC', reg_ret['IH+IC']), ('IF+IC', reg_ret['IF+IC']),
                  ('IH', single[0][1]), ('IF', single[1][1]), ('IC', single[2][1])], base))
plot_nav('Fig9 rotation vs single', {'A500': base, 'rotation': ret_mix, 'IH': single[0][1],
                                     'IF': single[1][1], 'IC': single[2][1]}, sm.map(sigmap))
plot_nav('Fig10 rotation vs combos', {'A500': base, 'rotation': ret_mix, 'IF+IC': reg_ret['IF+IC'],
                                      'IH+IC': reg_ret['IH+IC']}, sm.map(sigmap))

# ============================ 表8 / 图11: 叠加合约(基差)轮动 ============================
ret_mix_rot = (R['A500'] - pick_ret(sig_mix, ROT[HEDGE_PRODS])).loc[E]
print('\n' + '=' * 30 + ' 表8 风格轮动 + 合约轮动 ' + '=' * 30)
print(make_table([('IH+IF+IC', ret_mix), ('IH+IF+IC+合约轮动', ret_mix_rot)], base))
plot_nav('Fig11 rotation + contract rotation', {'A500': base, 'product rotation': ret_mix, 'product+contract rotation': ret_mix_rot})

print('\n注: 图1/图2(同花顺基金规模与私募发行量)无聚宽数据源，未复现；回测不含交易成本，历史回测有失效风险。')
```
