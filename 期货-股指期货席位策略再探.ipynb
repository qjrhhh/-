```python
# -*- coding: utf-8 -*-
"""
复现华泰期货研报《股指期货席位策略再探》(2025-07-29) —— 聚宽研究环境版本
环境: Python 3.6.7 / numpy 1.19.5 / pandas 0.23.4 / scipy 1.2.0 (旧写法, 无 f-string / to_numpy / panel)
数据: 聚宽 get_all_securities / get_price / finance.FUT_MEMBER_POSITION_RANK (龙虎榜)
说明: 不读写磁盘, 数据直接使用, 结果直接 print (图用 plt.show() 显示, 不保存)

复现内容:
  图1-4  持仓占比      | 表1  净持仓跟踪      | 表2  价格趋势
  表3    量价齐趋      | 表4  量价背离        | 表5  多空持仓(蜘蛛网)
  表6    持仓行情共振1 | 表7  持仓行情共振2   | 摘要: 各品种择优
"""
import datetime
import re
import warnings

import numpy as np
import pandas as pd
from jqdata import *
from jqdata import finance
try:
    from jqdata import query
except ImportError:
    pass

warnings.filterwarnings('ignore')
pd.set_option('display.unicode.east_asian_width', True)
pd.set_option('display.width', 220)
pd.set_option('display.max_rows', 500)
pd.set_option('display.max_columns', 50)

# ============================== 参数 ==============================
START_DATE = '2024-01-01'
END_DATE = '2026-12-20'          # 晚于今天时自动截断到今天
PRODUCTS = ['IH', 'IF', 'IC', 'IM']
TOPN = [5, 10, 15, 20]           # 跟踪席位排名 <= n
ANN = 242                        # 年化交易日
DRAW = True                      # 是否 plt.show() 画图 (不保存)
BIG_CAP = ['IH', 'IF']           # 大盘蓝筹: 趋势跟踪; IC/IM: 反转
RANK_LONG, RANK_SHORT = 501002, 501003   # 持买单量排名 / 持卖单量排名

_today = datetime.date.today()
_start = pd.to_datetime(START_DATE).date()
_end = min(pd.to_datetime(END_DATE).date(), _today)
print('回测区间: %s ~ %s (END_DATE 超过今天则截断到今天)' % (_start, _end))

# ============================== 交易日历 ==============================
_buf = _start - datetime.timedelta(days=30)
cal = pd.to_datetime([str(d) for d in get_trade_days(start_date=_buf, end_date=_end)])
prev_map = dict(zip(cal[1:], cal[:-1]))
next_map = dict(zip(cal[:-1], cal[1:]))
eval_days = cal[cal >= pd.Timestamp(_start)]

ALLSEC = get_all_securities(['futures'])


# ============================== 当季合约 ==============================
def quarter_contracts(prod):
    """某品种所有季月(3/6/9/12)合约及其起止日"""
    pat = re.compile(r'^%s(\d{2})(\d{2})\.CCFX$' % prod)
    rows = []
    for code in ALLSEC.index:
        m = pat.match(code)
        if m and int(m.group(2)) % 3 == 0:
            rows.append((code,
                         pd.Timestamp(ALLSEC.loc[code, 'start_date']),
                         pd.Timestamp(ALLSEC.loc[code, 'end_date'])))
    return pd.DataFrame(rows, columns=['code', 'start', 'end'])


def build_qc(prod):
    """每个交易日 u 的当季合约: 最近且尚未到期(end > u)的季月合约 (到期日当天已换到下一个当季)"""
    qdf = quarter_contracts(prod)
    res = {}
    for u in eval_days:
        c = qdf[(qdf['start'] <= u) & (qdf['end'] > u)]
        if len(c) > 0:
            res[u] = c.sort_values('end').iloc[0]['code']
    return pd.Series(res)


# ============================== 数据获取 ==============================
def fetch_px(code, s, e):
    df = get_price(code, start_date=s.strftime('%Y-%m-%d'), end_date=e.strftime('%Y-%m-%d'),
                   frequency='daily', fields=['open', 'close', 'open_interest'])
    df.index = pd.to_datetime(df.index)
    return df


def fetch_rank(code, s, e):
    """龙虎榜多空前20席位持仓; 接口单次最多4000行, 按90天切片查询"""
    T = finance.FUT_MEMBER_POSITION_RANK
    frames = []
    cur = s
    while cur <= e:
        nxt = min(cur + datetime.timedelta(days=89), e)
        q = query(T.day, T.rank, T.rank_type_ID, T.indicator).filter(
            T.code == code,
            T.rank_type_ID.in_([RANK_LONG, RANK_SHORT]),
            T.day >= cur, T.day <= nxt)
        df = finance.run_query(q)
        if df is not None and len(df) > 0:
            frames.append(df)
        cur = nxt + datetime.timedelta(days=1)
    if len(frames) == 0:
        return None
    return pd.concat(frames)


def topn_sums(rk):
    """前n席位 多头持仓合计 L{n}、空头持仓合计 S{n}"""
    rk = rk.copy()
    rk['day'] = pd.to_datetime(rk['day'])
    rk['indicator'] = rk['indicator'].astype(float)
    cols = {}
    for n in TOPN:
        sub = rk[rk['rank'] <= n]
        p = sub.groupby(['day', 'rank_type_ID'])['indicator'].sum().unstack()
        if RANK_LONG in p.columns:
            cols['L%d' % n] = p[RANK_LONG]
        if RANK_SHORT in p.columns:
            cols['S%d' % n] = p[RANK_SHORT]
    return pd.DataFrame(cols)


def build_contract_frame(code, win_s, win_e):
    px = fetch_px(code, win_s, win_e)
    rk = fetch_rank(code, win_s.date(), win_e.date())
    D = px.copy()
    if rk is not None:
        D = D.join(topn_sums(rk))
    for n in TOPN:
        for k in ('L', 'S'):
            if '%s%d' % (k, n) not in D.columns:
                D['%s%d' % (k, n)] = np.nan
    D['dprice'] = D['close'].pct_change()
    for n in TOPN:
        D['dnet%d' % n] = (D['L%d' % n] - D['S%d' % n]).diff()   # 净持仓变化
        D['dL%d' % n] = D['L%d' % n].diff()                      # 多头持仓变化
        D['dS%d' % n] = D['S%d' % n].diff()                      # 空头持仓变化
        D['rL%d' % n] = D['L%d' % n] / D['open_interest']        # 多头占比
        D['rS%d' % n] = D['S%d' % n] / D['open_interest']        # 空头占比
    return D


def build_features(prod):
    """
    行索引 u = 持仓日。
    因子取自 u 的前一交易日 t (t日收盘后得到席位数据 -> t+1 开盘交易), 且取自 u 日对应的当季合约;
    ret = 当季合约 u 开盘 -> u+1 开盘 收益 (t+1早上操作, t+2早上切换)。
    """
    qc = build_qc(prod)
    if len(qc) == 0:
        return None
    parts = []
    for c in qc.unique():
        us = qc[qc == c].index
        p1 = prev_map.get(us[0], us[0])
        p2 = prev_map.get(p1, p1)
        win_e = next_map.get(us[-1], us[-1])
        print('  %s 当季合约 %s: %s ~ %s' % (prod, c, us[0].date(), us[-1].date()))
        D = build_contract_frame(c, p2, win_e)
        ts = [prev_map.get(u, u) for u in us]
        sub = D.reindex(ts)
        sub.index = us
        nx = [next_map.get(u, u) for u in us]
        o_u = D['open'].reindex(us).values
        o_n = D['open'].reindex(nx).values
        ret = o_n / o_u - 1.0
        last_flag = np.array([a == b for a, b in zip(nx, us)])
        ret[last_flag] = np.nan
        sub['ret'] = ret
        sub['contract'] = c
        parts.append(sub)
    return pd.concat(parts).sort_index()


# ============================== 信号 ==============================
def mk_sig(index, long_c, short_c):
    """多=1, 空=-1, 其余 NaN (后续 ffill = 维持上一持仓)"""
    s = pd.Series(np.nan, index=index)
    s[long_c.values] = 1.0
    s[short_c.values] = -1.0
    return s


def sig_net(F, n, prod):                         # 净持仓跟踪
    d = F['dnet%d' % n]
    return mk_sig(F.index, d > 0, d < 0)


def sig_price(F, n, prod):                       # 价格趋势
    p = F['dprice']
    return mk_sig(F.index, p > 0, p < 0)


def sig_resonance(F, n, prod):                   # 量价齐趋
    d, p = F['dnet%d' % n], F['dprice']
    return mk_sig(F.index, (d > 0) & (p > 0), (d < 0) & (p < 0))


def sig_diverge(F, n, prod):                     # 量价背离
    d, p = F['dnet%d' % n], F['dprice']
    if prod == 'IC':                             # IC: 背离时跟随价格方向
        return mk_sig(F.index, (d < 0) & (p > 0), (d > 0) & (p < 0))
    return mk_sig(F.index, (d > 0) & (p < 0), (d < 0) & (p > 0))   # IH/IF/IM: 跟随净持仓方向


def sig_spider(F, n, prod):                      # 蜘蛛网(多空持仓)
    dl, ds = F['dL%d' % n], F['dS%d' % n]
    return mk_sig(F.index, (dl > 0) & (ds < 0), (dl < 0) & (ds > 0))


def sig_spider_p1(F, n, prod):                   # 持仓行情共振1: 蜘蛛网信号 + 价格同向
    dl, ds, p = F['dL%d' % n], F['dS%d' % n], F['dprice']
    return mk_sig(F.index, (dl > 0) & (ds < 0) & (p > 0), (dl < 0) & (ds > 0) & (p < 0))


def sig_spider_p2(F, n, prod):                   # 持仓行情共振2: 多空同增同减 + 价格
    dl, ds, p = F['dL%d' % n], F['dS%d' % n], F['dprice']
    up = (dl > ds) & (ds > 0) & (p > 0)
    dn = (dl < ds) & (ds < 0) & (p < 0)
    if prod in BIG_CAP:
        return mk_sig(F.index, up, dn)
    return mk_sig(F.index, dn, up)               # IC/IM 反向


# (名称, 对应研报表, 信号函数, 是否按席位排名n循环)
STRATEGIES = [
    ('净持仓跟踪',     '表1', sig_net,       True),
    ('价格趋势',       '表2', sig_price,     False),
    ('量价齐趋',       '表3', sig_resonance, True),
    ('量价背离',       '表4', sig_diverge,   True),
    ('多空持仓(蜘蛛网)', '表5', sig_spider,    True),
    ('持仓行情共振1',   '表6', sig_spider_p1, True),
    ('持仓行情共振2',   '表7', sig_spider_p2, True),
]


# ============================== 回测 & 指标 ==============================
def backtest(F, sig):
    pos = sig.ffill().fillna(0.0)
    ret = (pos * F['ret']).dropna()
    pos = pos.reindex(ret.index)
    return pos, ret


def perf(ret, pos):
    nv = (1.0 + ret).cumprod()
    ann = ret.mean() * ANN
    vol = ret.std() * np.sqrt(ANN)
    sharpe = ann / vol if vol > 0 else np.nan
    mdd = (nv / nv.cummax() - 1.0).min()
    active = pos != 0
    win = (ret[active] > 0).mean() if active.sum() > 0 else np.nan
    prev = pos.shift(1)
    nsig = int(((pos != prev) & prev.notnull() & (prev != 0)).sum())     # 方向切换次数
    freq = nsig / float(len(ret)) if len(ret) > 0 else np.nan
    return nv, [ann, vol, sharpe, mdd, win, nsig, freq]


def fmt_table(rows):
    cols = ['品种', '跟踪席位排名(<=n)', '年化收益', '年化波动率', '夏普比率', '最大回撤', '胜率', '信号次数', '信号频率']
    out = []
    for r in rows:
        out.append([r[0], r[1],
                    '%.2f%%' % (r[2] * 100), '%.2f%%' % (r[3] * 100), '%.2f' % r[4],
                    '%.2f%%' % (r[5] * 100), '%.2f%%' % (r[6] * 100), r[7], '%.2f%%' % (r[8] * 100)])
    return pd.DataFrame(out, columns=cols)


def plot_group(title, prod_nv):
    import matplotlib.pyplot as plt
    fig, axes = plt.subplots(2, 2, figsize=(14, 8))
    for ax, prod in zip(axes.flat, PRODUCTS):
        d = prod_nv.get(prod)
        if not d:
            continue
        for lab, nv in d.items():
            ax.plot(nv.index, nv.values, label=str(lab))
        ax.set_title(prod)
        ax.legend(fontsize=8, ncol=3)
        ax.grid(True, alpha=0.3)
    fig.suptitle(title)
    plt.show()


# ============================== 主流程 ==============================
print('\n>>> 获取数据 (当季合约 + 龙虎榜) ...')
FEATS = {}
for prod in PRODUCTS:
    print('[%s]' % prod)
    f = build_features(prod)
    if f is None or len(f) == 0:
        print('  %s 区间内无可用合约, 跳过' % prod)
        continue
    FEATS[prod] = f
print('数据获取完成: ' + ', '.join(['%s %d个交易日' % (k, len(v)) for k, v in FEATS.items()]))

# ---------- 图1-4: 持仓占比 ----------
print('\n' + '=' * 30 + ' 图1-4 前n席位持仓占当季合约总持仓比例 (区间均值 / 最新) ' + '=' * 30)
ratio_rows = []
ratio_plot = {}
for prod, F in FEATS.items():
    ratio_plot[prod] = {}
    for n in TOPN:
        rl, rs = F['rL%d' % n].dropna(), F['rS%d' % n].dropna()
        if len(rl) == 0:
            continue
        ratio_rows.append([prod, n, '%.2f%%' % (rl.mean() * 100), '%.2f%%' % (rs.mean() * 100),
                           '%.2f%%' % (rl.iloc[-1] * 100), '%.2f%%' % (rs.iloc[-1] * 100)])
        ratio_plot[prod]['Long_%d' % n] = rl
        ratio_plot[prod]['Short_%d' % n] = rs
print(pd.DataFrame(ratio_rows, columns=['品种', '前n席位', '多头占比均值', '空头占比均值', '多头占比最新', '空头占比最新']).to_string(index=False))
if DRAW:
    plot_group('Fig1-4 top-n seats position share', ratio_plot)

# ---------- 表1-7 ----------
ALL_ROWS = []     # 用于择优
for name, tab, fn, loop_n in STRATEGIES:
    rows = []
    prod_nv = {}
    for prod, F in FEATS.items():
        prod_nv[prod] = {}
        ns = TOPN if loop_n else [None]
        for n in ns:
            sig = fn(F, n if n else TOPN[0], prod)
            pos, ret = backtest(F, sig)
            if len(ret) < 5:
                continue
            nv, m = perf(ret, pos)
            lab = n if n else '-'
            rows.append([prod, lab] + m)
            ALL_ROWS.append([name, tab, prod, lab] + m)
            prod_nv[prod]['top%s' % lab if n else 'price'] = nv
        # 基准: 当季合约多头滚动持有
        bpos = pd.Series(1.0, index=F.index)
        bpos, bret = backtest(F, bpos)
        bnv, bm = perf(bret, bpos)
        rows.append([prod, '-'] + bm)
        prod_nv[prod]['benchmark'] = bnv
    print('\n' + '=' * 30 + ' %s 策略回测结果 (研报%s) ' % (name, tab) + '=' * 30)
    print(fmt_table(rows).to_string(index=False))
    print('注: 基准策略为当季合约多头滚动持有, 跟踪席位排名记为"-", 未计交易成本。')
    if DRAW:
        plot_group('%s net value' % name, prod_nv)

# ---------- 摘要: 各品种择优 ----------
print('\n' + '=' * 30 + ' 各品种不同策略择优 (按夏普比率) ' + '=' * 30)
cols = ['策略', '研报表', '品种', '跟踪席位排名(<=n)', 'ann', 'vol', 'sharpe', 'mdd', 'win', 'nsig', 'freq']
AR = pd.DataFrame(ALL_ROWS, columns=cols)
best = []
for prod in PRODUCTS:
    sub = AR[AR['品种'] == prod]
    sub = sub[sub['sharpe'].notnull()]
    if len(sub) == 0:
        continue
    r = sub.sort_values('sharpe', ascending=False).iloc[0]
    best.append([prod, r['策略'], r['跟踪席位排名(<=n)'],
                 '%.2f%%' % (r['ann'] * 100), '%.2f' % r['sharpe'], '%.2f%%' % (r['mdd'] * 100)])
print(pd.DataFrame(best, columns=['品种', '最优策略', '跟踪席位排名(<=n)', '年化收益', '夏普比率', '最大回撤']).to_string(index=False))
print('\n风险提示: 回测基于历史数据, 有失效风险。')
```
