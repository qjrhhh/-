```python
# -*- coding: utf-8 -*-
"""
华泰期货《偏而不倚：偏度因子策略再研究》复现 —— 聚宽研究环境版
- 数据全部来自聚宽 get_price，直接使用，不落盘
- 结果全部 print / plt.show() 显示，不写磁盘
- 写法偏旧：不使用 f-string / 类型注解 / pd.append / panel
- 回测区间：BT_START ~ BT_END（为计算历史黑名单、动量等，数据向前多取到 DATA_START，
  同时会输出研报口径(2010至今/2014至今)的全样本结果做对照）
"""
import datetime
import warnings
from collections import Counter

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

warnings.filterwarnings('ignore')
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False
pd.set_option('display.width', 250)
pd.set_option('display.max_columns', 30)

# ============================ 参数 ============================
DATA_START = '2009-01-01'      # 数据起点（给150日偏度/250日动量/历史黑名单预热）
REPORT_START = '2010-01-01'    # 研报全样本起点
BT_START = '2024-01-01'        # 指定回测区间
BT_END = '2026-12-20'
SKEW_WIN = 150                 # 偏度窗口
N_SEL = 5                      # 多空各选5个
LIQ_Q = 0.2                    # 流动性过滤：近60日均成交额排名后20%不参与
MOM_LONG, MOM_SHORT, VOL_WIN = 250, 20, 250
W_MULTI = (0.5, 0.25, 0.25)    # 多因子权重：(偏度, 动量, 波动率)，研报未披露，可自行调整
BL_CUM, BL_CNT = -0.015, 3     # 九月黑名单：累计贡献<-1.5% 或 负收益次数>=3

# 商品期货品种（聚宽字典中的商品品种，已剔除已退市旧代码和极不活跃品种）
PRODUCTS = [
    ('A', 'XDCE', '豆一'), ('B', 'XDCE', '豆二'), ('M', 'XDCE', '豆粕'), ('C', 'XDCE', '玉米'),
    ('Y', 'XDCE', '豆油'), ('L', 'XDCE', '聚乙烯'), ('P', 'XDCE', '棕榈油'), ('V', 'XDCE', 'PVC'),
    ('J', 'XDCE', '焦炭'), ('JM', 'XDCE', '焦煤'), ('I', 'XDCE', '铁矿石'), ('JD', 'XDCE', '鸡蛋'),
    ('PP', 'XDCE', '聚丙烯'), ('CS', 'XDCE', '玉米淀粉'), ('EG', 'XDCE', '乙二醇'),
    ('EB', 'XDCE', '苯乙烯'), ('PG', 'XDCE', '液化石油气'), ('LH', 'XDCE', '生猪'),
    ('SC', 'XINE', '原油'), ('NR', 'XINE', '20号胶'), ('LU', 'XINE', '低硫燃料油'),
    ('BC', 'XINE', '国际铜'),
    ('AL', 'XSGE', '铝'), ('RU', 'XSGE', '天然橡胶'), ('FU', 'XSGE', '燃料油'), ('CU', 'XSGE', '铜'),
    ('ZN', 'XSGE', '锌'), ('AU', 'XSGE', '黄金'), ('RB', 'XSGE', '螺纹钢'), ('PB', 'XSGE', '铅'),
    ('AG', 'XSGE', '白银'), ('BU', 'XSGE', '沥青'), ('HC', 'XSGE', '热轧卷板'), ('SN', 'XSGE', '锡'),
    ('NI', 'XSGE', '镍'), ('SP', 'XSGE', '纸浆'), ('SS', 'XSGE', '不锈钢'), ('BR', 'XSGE', '丁二烯橡胶'),
    ('CF', 'XZCE', '棉花'), ('SR', 'XZCE', '白糖'), ('TA', 'XZCE', 'PTA'), ('OI', 'XZCE', '菜油'),
    ('FG', 'XZCE', '玻璃'), ('RM', 'XZCE', '菜粕'), ('MA', 'XZCE', '甲醇'), ('SF', 'XZCE', '硅铁'),
    ('SM', 'XZCE', '锰硅'), ('ZC', 'XZCE', '动力煤'), ('CY', 'XZCE', '棉纱'), ('AP', 'XZCE', '苹果'),
    ('UR', 'XZCE', '尿素'), ('SA', 'XZCE', '纯碱'), ('PF', 'XZCE', '短纤'), ('PK', 'XZCE', '花生'),
    ('SH', 'XZCE', '烧碱'), ('PX', 'XZCE', '对二甲苯'),
    ('SI', 'XGFE', '工业硅'), ('LC', 'XGFE', '碳酸锂'),
]
EXMAP = {'XGFE': 'GFEX'}   # 广期所代码后缀为 GFEX


def code_of(sym, ex, tail):
    return '{}{}.{}'.format(sym, tail, EXMAP.get(ex, ex))


# ============================ 取数 ============================
def load_all():
    end = min(pd.Timestamp(BT_END), pd.Timestamp(datetime.date.today()))
    end_s = end.strftime('%Y-%m-%d')
    c8, c9, mo, names = {}, {}, {}, {}
    for sym, ex, name in PRODUCTS:
        k8 = code_of(sym, ex, '8888')
        k9 = code_of(sym, ex, '9999')
        try:
            d8 = get_price(k8, start_date=DATA_START, end_date=end_s, frequency='daily',
                           fields=['close', 'money'], skip_paused=False)
            d9 = get_price(k9, start_date=DATA_START, end_date=end_s, frequency='daily',
                           fields=['close'], skip_paused=False)
        except Exception as e:
            print('跳过 {} {}: {}'.format(k9, name, e))
            continue
        if d8 is None or d9 is None:
            continue
        d8 = d8.dropna(subset=['close'])
        if len(d8) < 250:
            print('数据不足，跳过', name)
            continue
        c8[k9] = d8['close']
        mo[k9] = d8['money']
        c9[k9] = d9['close'].dropna()
        names[k9] = name
    C8 = pd.DataFrame(c8)
    C9 = pd.DataFrame(c9)
    M = pd.DataFrame(mo)
    keep = C8.notna().sum(axis=1) >= 5          # 去掉非交易日
    C8, C9, M = C8[keep], C9.reindex(C8[keep].index), M.reindex(C8[keep].index)
    return C8.sort_index(), C9.sort_index(), M.sort_index(), names


print('正在取数……')
C8, C9, MONEY, NAME = load_all()
C8 = C8.ffill(limit=5)
C9 = C9.ffill(limit=5)
R = C8 / C8.shift(1) - 1.0                      # 日收益率：品种指数(8888，持仓加权，避免主力换月跳空)
DATES = R.index
COLS = list(R.columns)
print('品种数: {}, 日期: {} ~ {}'.format(len(COLS), DATES[0].date(), DATES[-1].date()))

# ============================ 因子 ============================


def pop_skew(r, win, minp):
    m1 = r.rolling(win, min_periods=minp).mean()
    m2 = (r ** 2).rolling(win, min_periods=minp).mean()
    m3 = (r ** 3).rolling(win, min_periods=minp).mean()
    var = m2 - m1 ** 2
    num = m3 - 3 * m1 * m2 + 2 * m1 ** 3
    return num / (var ** 1.5)


SKEW = pop_skew(R, SKEW_WIN, 120)
MOM_L = C8 / C8.shift(MOM_LONG) - 1.0
MOM_S = C8 / C8.shift(MOM_SHORT) - 1.0
VOL = R.rolling(VOL_WIN, min_periods=200).std()
TERM = C9 / C8 - 1.0      # 期限结构代理：主力价/持仓加权指数价，>0 近似现货升水(back)，<0 近似contango

# 流动性 & 样本充足过滤
cnt_ok = R.notna().astype(float).rolling(SKEW_WIN).sum() >= 120
money_avg = MONEY.rolling(60, min_periods=40).mean()
liq_ok = money_avg.rank(axis=1, pct=True) >= LIQ_Q
ELIG = cnt_ok & liq_ok

# 月末调仓日
ME_DATES = DATES.to_series().groupby([DATES.year, DATES.month]).max().values
ME_DATES = pd.DatetimeIndex(ME_DATES)


# ============================ 回测引擎 ============================
def run_strategy(S, n=N_SEL, blacklist=None):
    """S: 日期x品种得分，得分越高越做多。月末信号，次日起持仓一个月，等权不加杠杆。"""
    Wme = pd.DataFrame(0.0, index=ME_DATES, columns=COLS)
    for d in ME_DATES:
        s = S.loc[d].where(ELIG.loc[d]).dropna()
        nm = d.month % 12 + 1
        ny = d.year + (1 if d.month == 12 else 0)
        if blacklist is not None and nm == 9:
            bad = [a for a in blacklist.get(ny, []) if a in s.index]
            s = s.drop(bad)
        if len(s) < 2 * n:
            continue
        Wme.loc[d, s.nlargest(n).index] = 1.0 / (2 * n)
        Wme.loc[d, s.nsmallest(n).index] = -1.0 / (2 * n)
    W = Wme.reindex(DATES).ffill().shift(1).fillna(0.0)
    C = W * R.fillna(0.0)
    C = C[C.index >= REPORT_START]
    Wc = W[W.index >= REPORT_START]
    ret = C.sum(axis=1)
    lg = C.where(Wc > 0, 0.0).sum(axis=1)
    sh = C.where(Wc < 0, 0.0).sum(axis=1)
    return {'Wme': Wme, 'W': Wc, 'C': C, 'ret': ret, 'long': lg, 'short': sh}


def make_blacklist(C):
    sept = C[C.index.month == 9]
    ys = sept.groupby(sept.index.year).sum()
    bl = {}
    for Y in range(2011, DATES[-1].year + 1):
        hist = ys[ys.index < Y]
        if len(hist) == 0:
            continue
        cum = hist.sum()
        neg = (hist < 0).sum()
        bl[Y] = [c for c in COLS if (cum[c] < BL_CUM) or (neg[c] >= BL_CNT)]
    return bl


def rank_score(F, ascending_best_low=False):
    """行内百分位排名，越大越好；F已经做过ELIG掩码"""
    return F.rank(axis=1, pct=True)


print('回测中……')
res_skew = run_strategy(-SKEW)
res_momL = run_strategy(MOM_L)
res_momS = run_strategy(MOM_S)
res_term = run_strategy(TERM)
BL = make_blacklist(res_skew['C'])
res_bl = run_strategy(-SKEW, blacklist=BL)

# 多因子：偏度(低优先) + 动量(强优先) + 波动率(低优先)，各自截面排名后加权
mask = ELIG
r_sk = rank_score((-SKEW).where(mask))
r_mo = rank_score(MOM_L.where(mask))
r_vo = rank_score((-VOL).where(mask))
MULTI = W_MULTI[0] * r_sk + W_MULTI[1] * r_mo + W_MULTI[2] * r_vo
res_multi = run_strategy(MULTI, blacklist=BL)

BENCH = R.mean(axis=1)    # 等权商品指数（南华商品指数的代理）


# ============================ 工具函数 ============================
def calc_stats(r):
    r = r.dropna()
    if len(r) < 30:
        return None
    nav = (1 + r).cumprod()
    ann = nav.iloc[-1] ** (252.0 / len(r)) - 1
    vol = r.std() * np.sqrt(252)
    var = np.percentile(r, 5)
    cvar = r[r <= var].mean()
    mdd = (nav / nav.cummax() - 1).min()
    return [ann, vol, r.kurt(), r.skew(), cvar, var, mdd, r.mean() * 252 / vol, ann / vol]


def show_stats(title, ret):
    periods = [('2010至今', ret[ret.index >= REPORT_START]),
               ('2014至今', ret[ret.index >= '2014-01-01']),
               ('{}~{}'.format(BT_START, BT_END), ret[(ret.index >= BT_START) & (ret.index <= BT_END)])]
    rows, idx = [], []
    for nm, r in periods:
        s = calc_stats(r)
        if s is None:
            continue
        idx.append(nm)
        rows.append(['{:.2f}%'.format(s[0] * 100), '{:.2f}%'.format(s[1] * 100), '{:.2f}'.format(s[2]),
                     '{:.2f}'.format(s[3]), '{:.2f}%'.format(s[4] * 100), '{:.2f}%'.format(s[5] * 100),
                     '{:.2f}%'.format(s[6] * 100), '{:.2f}'.format(s[7]), '{:.2f}'.format(s[8])])
    df = pd.DataFrame(rows, index=idx, columns=['年化收益', '年化波动', '峰度', '偏度', 'CVar', 'Var',
                                               '最大回撤', '夏普比率', '信息比率'])
    print('\n' + title)
    print(df.to_string())


def hist_den(ax, x, bins=40, color='#C00000'):
    try:
        ax.hist(x, bins=bins, density=True, color=color, edgecolor='white')
    except TypeError:
        ax.hist(x, bins=bins, normed=True, color=color, edgecolor='white')


def plot_navs(rets, title, start=None):
    plt.figure(figsize=(11, 4.5))
    for nm, r, col in rets:
        rr = r if start is None else r[(r.index >= start) & (r.index <= BT_END)]
        plt.plot((1 + rr).cumprod().index, (1 + rr).cumprod().values, label=nm, color=col, linewidth=1.3)
    plt.title(title)
    plt.legend(loc='upper left')
    plt.grid(alpha=0.3)
    plt.show()


def year_prod(r):
    return (1 + r).groupby(r.index.year).prod() - 1


def names_of(lst):
    return [NAME[c] for c in lst]


# ============================ 图1 四大因子净值 ============================
plot_navs([('偏度', res_skew['ret'], '#C00000'), ('长周期动量', res_momL['ret'], 'black'),
           ('短周期动量', res_momS['ret'], 'gray'), ('期限结构(代理)', res_term['ret'], '#F4B183')],
          '图1: 商品策略指数历史净值(研报区间)')
plot_navs([('偏度', res_skew['ret'], '#C00000'), ('长周期动量', res_momL['ret'], 'black'),
           ('短周期动量', res_momS['ret'], 'gray'), ('期限结构(代理)', res_term['ret'], '#F4B183')],
          '图1(补): 指定回测区间净值 {}~{}'.format(BT_START, BT_END), start=BT_START)

# ============================ 图2 多空收益 ============================
yl, ys_, yc = year_prod(res_skew['long']), year_prod(res_skew['short']), year_prod(res_skew['ret'])
yrs = yc.index.values
plt.figure(figsize=(12, 4.5))
plt.bar(yrs - 0.2, yl.values * 100, width=0.4, color='#C00000', label='多头收益率')
plt.bar(yrs + 0.2, ys_.values * 100, width=0.4, color='#222222', label='空头收益率')
plt.plot(yrs, yc.values * 100, color='gray', linewidth=2, label='组合收益率')
plt.title('图2: 偏度策略多空收益一览(%)')
plt.legend()
plt.grid(alpha=0.3)
plt.show()

# ============================ 图3 月度收益 vs 商品指数月度波动率 ============================
ret0 = res_skew['ret']
ym = ret0.index.to_period('M')
mret = (1 + ret0).groupby(ym).prod() - 1
mvol = BENCH[BENCH.index >= REPORT_START].groupby(BENCH[BENCH.index >= REPORT_START].index.to_period('M')).std()
fig, ax1 = plt.subplots(figsize=(13, 4.5))
xs = [p.to_timestamp() for p in mret.index]
ax1.bar(xs, mret.values, width=20, color='#C00000', label='偏度策略月度收益率')
ax2 = ax1.twinx()
ax2.plot([p.to_timestamp() for p in mvol.index], mvol.values, color='silver', label='等权商品指数月度波动率(右轴, 南华指数代理)')
ax1.set_title('图3: 偏度策略月度收益拆解')
ax1.legend(loc='upper left')
ax2.legend(loc='upper right')
plt.show()

# ============================ 表1/表2 年度多空持仓前三 ============================


def top3_table(Wme, side):
    cnt = {}
    for d in Wme.index:
        row = Wme.loc[d]
        if row.abs().sum() == 0:
            continue
        y = d.year + 1 if d.month == 12 else d.year
        sel = list(row[row > 0].index) if side == 'long' else list(row[row < 0].index)
        cnt.setdefault(y, Counter()).update(sel)
    rows = []
    for y in sorted(cnt):
        if y < 2010:
            continue
        t = cnt[y].most_common(3)
        row = [y]
        for c, n in t:
            row.append(NAME[c])
        for c, n in t:
            row.append(n)
        rows.append(row)
    return pd.DataFrame(rows, columns=['年份', '品种一', '品种二', '品种三', '频次一', '频次二', '频次三'])


print('\n表1: 偏度策略年度空头持仓前三名品种')
print(top3_table(res_skew['Wme'], 'short').to_string(index=False))
print('\n表2: 偏度策略年度多头持仓前三名品种')
print(top3_table(res_skew['Wme'], 'long').to_string(index=False))

# ============================ 图4 品种年度贡献热力图 ============================
A = res_skew['C'].groupby(res_skew['C'].index.year).sum() * 100
A = A.loc[:, (A.abs().sum() > 0)]
A.columns = [NAME[c] for c in A.columns]
vmax = np.nanmax(np.abs(A.values))
plt.figure(figsize=(12, 0.22 * A.shape[1] + 2))
plt.imshow(A.T.values, aspect='auto', cmap='RdBu_r', vmin=-vmax, vmax=vmax)
plt.xticks(range(len(A.index)), A.index, rotation=90, fontsize=8)
plt.yticks(range(A.shape[1]), A.columns, fontsize=7)
plt.colorbar()
plt.title('图4: 偏度策略各品种年度贡献(%)')
plt.show()

# ============================ 图5 年度收益-波动散点 ============================
yvol = ret0.groupby(ret0.index.year).std() * np.sqrt(252)
yann = year_prod(ret0)
plt.figure(figsize=(7, 5))
plt.scatter(yvol.values, yann.values, color='black')
for y in yvol.index:
    plt.annotate(str(y), (yvol[y], yann[y]), fontsize=8)
k, b = np.polyfit(yvol.values, yann.values, 1)
xx = np.linspace(yvol.min(), yvol.max(), 50)
plt.plot(xx, k * xx + b, color='red')
plt.xlabel('年度波动率(标准差)')
plt.ylabel('年度收益率')
plt.title('图5: 偏度策略收益波动率对比一览')
plt.grid(alpha=0.3)
plt.show()

# ============================ 表3 ============================
show_stats('表3: 偏度策略盈亏统计', res_skew['ret'])

# ============================ 图6 偏度示意图 ============================
x = np.linspace(-4, 4, 400)
fig, axs = plt.subplots(1, 2, figsize=(11, 3.5))
for ax, a, t in zip(axs, [-5, 5], ['Negative Skew', 'Positive Skew']):
    try:
        ax.plot(x, stats.skewnorm.pdf(x, a, loc=-a * 0.15, scale=1.2), color='red')
    except Exception:
        pass
    ax.plot(x, stats.norm.pdf(x, 0, 1.2), '--', color='gray')
    ax.set_title(t)
plt.suptitle('图6: 收益率分布偏度示意图')
plt.show()

# ============================ 图7 最新一期偏度因子值 ============================
last = SKEW.index[-1]
sk_last = SKEW.loc[last].where(ELIG.loc[last]).dropna().sort_values()
sk_last.index = [NAME[c] for c in sk_last.index]
plt.figure(figsize=(7, 0.22 * len(sk_last) + 1.5))
plt.barh(range(len(sk_last)), sk_last.values, color='#C00000')
plt.yticks(range(len(sk_last)), sk_last.index, fontsize=8)
plt.title('图7: 最新一期偏度因子值 ({})'.format(last.date()))
plt.show()
print('偏度最高5个(做空):', list(sk_last.index[-5:][::-1]))
print('偏度最低5个(做多):', list(sk_last.index[:5]))

# ============================ 图8-15 典型案例 ============================


def case_plot(sym_ex, label, tag):
    code = sym_ex
    if code not in COLS:
        print('案例品种不在样本中:', label)
        return
    fig, axs = plt.subplots(2, 2, figsize=(13, 8))
    p = C9[code].dropna()
    axs[0, 0].plot(p.index, p.values, color='#C00000')
    axs[0, 0].set_title('{}主力合约走势'.format(label))
    sk = SKEW[code].dropna()
    axs[0, 1].plot(sk.index, sk.values, color='#C00000')
    axs[0, 1].set_title('{}偏度因子值走势'.format(label))
    r = R[code].dropna()
    r2 = r[r.index >= BT_START]
    if len(r2) < 60:
        r2 = r.iloc[-250:]
    xs_ = np.linspace(r2.min(), r2.max(), 300)
    kde = stats.gaussian_kde(r2.values)
    axs[1, 0].plot(xs_, kde(xs_), color='red', label=tag)
    axs[1, 0].plot(xs_, stats.norm.pdf(xs_, r2.mean(), r2.std()), '--', color='black', label='Normal')
    axs[1, 0].legend()
    axs[1, 0].set_title('{}主力收益率概率密度分布'.format(label))
    hist_den(axs[1, 1], r2.values)
    axs[1, 1].plot(xs_, stats.norm.pdf(xs_, r2.mean(), r2.std()), '--', color='black')
    axs[1, 1].set_title('{}主力收益率频数分布'.format(label))
    plt.suptitle('图8-11/12-15: {}案例'.format(label))
    plt.show()
    print('{} 日收益: 样本{}天, 偏度={:.2f}, 超额峰度={:.2f}'.format(label, len(r2), r2.skew(), r2.kurt()))


case_plot('HC9999.XSGE', '热轧卷板', 'HC')
case_plot('AG9999.XSGE', '白银', 'AG')

# ============================ 图16/17 月度季节性 ============================
mm = ret0.groupby(ret0.index.month)
m_ann = mm.mean() * 252
m_sharpe = mm.mean() / mm.std() * np.sqrt(252)
fig, axs = plt.subplots(1, 2, figsize=(12, 3.8))
axs[0].bar(m_ann.index, m_ann.values * 100, color='#C00000')
axs[0].set_title('图16: 偏度策略月均收益(年化,%)')
axs[1].plot(m_sharpe.index, m_sharpe.values, 'o-', color='#C00000')
axs[1].set_title('图17: 偏度策略月均夏普')
for ax in axs:
    ax.set_xticks(range(1, 13))
    ax.grid(alpha=0.3)
plt.show()
print('九月年化收益: {:.2f}%, 夏普: {:.2f}'.format(m_ann[9] * 100, m_sharpe[9]))

# ============================ 图18/19 箱线图 & t检验 ============================
plt.figure(figsize=(11, 4))
plt.boxplot([ret0[ret0.index.month == m].values for m in range(1, 13)])
plt.xticks(range(1, 13), range(1, 13))
plt.title('图18: 偏度策略月度收益(日度)分布盒状图')
plt.show()
sep = ret0[ret0.index.month == 9]
non = ret0[ret0.index.month != 9]
t_v, p_v = stats.ttest_ind(sep.values, non.values, equal_var=False)
print('\n图19: 九月负效应统计检验')
print(pd.DataFrame([['非九月', '{:.3f}%'.format(non.mean() * 100), '-', '-'],
                    ['九月', '{:.3f}%'.format(sep.mean() * 100), '{:.2f}'.format(t_v), '{:.3f}'.format(p_v)]],
                   columns=['月份类型', '日均收益', 'T统计量', 'P值']).to_string(index=False))

# ============================ 图20/21 信号强度 ============================
abs_sk = SKEW.where(ELIG).abs().mean(axis=1).dropna()
abs_sk = abs_sk[abs_sk.index >= REPORT_START]
mab = abs_sk.groupby(abs_sk.index.month).mean()
port_sk_sep, port_sk_non = [], []
Wm = res_skew['Wme']
for d in Wm.index:
    row = Wm.loc[d]
    if row.abs().sum() == 0 or d < pd.Timestamp(REPORT_START):
        continue
    lo = row[row > 0].index
    v = SKEW.loc[d, lo].mean()
    if d.month == 8:
        port_sk_sep.append(v)
    else:
        port_sk_non.append(v)
fig, axs = plt.subplots(1, 2, figsize=(12, 4))
axs[0].bar(mab.index, mab.values, color='#E00018')
axs[0].set_ylim(mab.min() * 0.9, mab.max() * 1.03)
axs[0].set_xticks(range(1, 13))
axs[0].set_title('图20: 偏度因子信号强度(平均|Skew|)月度分布')
axs[1].boxplot([port_sk_non, port_sk_sep])
axs[1].set_xticklabels(['非九月', '九月'])
axs[1].set_title('图21: 偏度强度对比(多头组合平均偏度)')
plt.show()

# ============================ 表4 九月持仓 ============================
rows = []
for d in Wm.index:
    if d.month == 8 and d.year >= 2010 and Wm.loc[d].abs().sum() > 0:
        row = Wm.loc[d]
        rows.append(['{}年9月'.format(d.year), ', '.join(names_of(row[row > 0].index)),
                     ', '.join(names_of(row[row < 0].index))])
print('\n表4: 偏度策略九月多空持仓一览')
print(pd.DataFrame(rows, columns=['年份', '多头持仓', '空头持仓']).to_string(index=False))

# ============================ 表5 九月黑名单 ============================
rows = [[y, ', '.join(names_of(BL[y]))] for y in sorted(BL) if y >= 2011 and y <= DATES[-1].year]
print('\n表5: 九月剔除品种池一览')
print(pd.DataFrame(rows, columns=['年份', '黑名单品类']).to_string(index=False))

# ============================ 图22 + 表6 ============================
plot_navs([('改善九月负效应偏度', res_bl['ret'], 'red'), ('原始偏度', res_skew['ret'], '#333333')],
          '图22: 改善九月负效应后策略净值对比')
plot_navs([('改善九月负效应偏度', res_bl['ret'], 'red'), ('原始偏度', res_skew['ret'], '#333333')],
          '图22(补): 指定区间 {}~{}'.format(BT_START, BT_END), start=BT_START)
show_stats('表6: 改善九月负效应后策略盈亏统计', res_bl['ret'])

# ============================ 图23 + 表7 ============================
plot_navs([('多因子', res_multi['ret'], 'red'), ('原始偏度', res_skew['ret'], '#333333'),
           ('改善九月负效应偏度', res_bl['ret'], 'gray')], '图23: 多因子融合策略净值对比')
plot_navs([('多因子', res_multi['ret'], 'red'), ('原始偏度', res_skew['ret'], '#333333'),
           ('改善九月负效应偏度', res_bl['ret'], 'gray')],
          '图23(补): 指定区间 {}~{}'.format(BT_START, BT_END), start=BT_START)
show_stats('表7: 多因子融合策略盈亏统计', res_multi['ret'])

# ============================ 指定区间当前持仓 ============================
for nm, res in [('原始偏度', res_skew), ('九月黑名单偏度', res_bl), ('多因子', res_multi)]:
    Wl = res['Wme']
    Wl = Wl[Wl.abs().sum(axis=1) > 0]
    d = Wl.index[-1]
    row = Wl.loc[d]
    print('\n[{}] 最近一次调仓日 {} 多头: {} | 空头: {}'.format(
        nm, d.date(), ', '.join(names_of(row[row > 0].index)), ', '.join(names_of(row[row < 0].index))))
print('\n完成')
```
