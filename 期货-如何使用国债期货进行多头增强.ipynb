```python
# -*- coding: utf-8 -*-
"""
华泰期货《如何使用国债期货进行多头增强》(2025-12-10) 研报复现 —— 聚宽研究环境版
------------------------------------------------------------------------------
* 兼容旧环境: Python 3.6.7 / numpy 1.19.5 / pandas 0.23.4 / scipy 1.2.0
  (不使用 f-string 以外的新语法, 不依赖 pywt / hmmlearn / filterpy, 全部用 numpy 手写)
* 数据: 全部通过聚宽 get_price / get_dominant_future / get_trade_days 在线获取, 直接使用, 不写磁盘
* 结果: 全部 print 输出 (可选 plt.show() 画图, 不落盘)
* 模型流程(与研报一致):
    卡尔曼滤波(时变对冲比率+新息) -> 小波高频分量(因果 à trous Haar) -> 数据驱动新息波动率预测(DDIVF)
    -> 隐马尔可夫(2 状态, 平稳/动荡) -> 状态相关阈值 P0/P1, 滚动窗口网格寻优
  策略 A: 多头替代增强 (价差过大 -> 国债ETF 临时换成国债期货多头)
  策略 B: 叠加策略   (价差过小 -> 做多价差[持ETF+空期货]; 价差过大 -> 多头替代; 正常 -> 持ETF)
"""
import bisect
import datetime as dt
import warnings

import numpy as np
import pandas as pd

warnings.filterwarnings('ignore')
try:
    from jqdata import *   # 聚宽研究环境里 get_price / get_dominant_future / get_trade_days 等
except Exception:
    pass

# ============================== 参数区 ==============================
START = dt.date(2024, 1, 1)          # 回测起点
END = dt.date(2026, 10, 20)          # 回测终点
WARM_START = dt.date(2023, 5, 8)     # 预热数据起点(用于卡尔曼/波动率/HMM 训练, TL 2023-04-21 上市)

# 研报: 511130 用 30 分钟线, 其余用 1 小时线; 期货端 TL(30年) 与 T(10年) 主力合约
CASES = [
    {'etf': '511130.XSHG', 'fut': 'TL', 'L': 30},
    {'etf': '511090.XSHG', 'fut': 'TL', 'L': 60},
    {'etf': '511260.XSHG', 'fut': 'T',  'L': 60},
]
TICK_FUT = {'TL': 0.01, 'T': 0.005}
TICK_ETF = 0.001
MULT = 10000.0                 # 国债期货: 面值100万 -> 价格每1点=1万元
INIT_CASH = 1e7                # 初始资金 1000 万
ETF_RATIO = 0.95               # 每次现货开仓使用 95% 资金
ETF_SLIP_TICKS = 3             # ETF 滑点 3 跳
FUT_SLIP_TICKS = 1             # 期货滑点 1 跳
ETF_FEE = 0.85e-4              # ETF 手续费 万0.85 (可被减免 -> 对比两种情形)
FUT_FEE = 3.0                  # 期货 3 元/手

TRAIN_DAYS = 120               # 滚动训练窗口 X (交易日)
STEP_DAYS = 20                 # 参数更新间隔 (交易日)
P_GRID = [0.5, 1.0, 1.5, 2.0, 2.5, 3.0]   # P0/P1 网格

INIT_N = 100                   # 卡尔曼 OLS 初始化长度(bar)
VALID_FROM = 220               # 前若干根 bar 的信号视为无效 (模型预热)
KF_Q = (1e-8, 1e-5)            # 卡尔曼状态噪声 [beta, alpha]
WAVE_LEVELS = 3                # 小波分解层数, 保留 1~3 层细节(高频)
MEAN_WIN = 100                 # 新息均值滚动窗口
PLOT = True                    # 是否 plt.show() 画净值图 (不落盘)


# ============================== 数据获取 ==============================
_ERR = [0]


def get_1m(code, d1, d2):
    """分段取 1 分钟 open/close (每段<=25 个自然日), 返回 DataFrame"""
    out = []
    cur = d1
    while cur <= d2:
        nxt = min(cur + dt.timedelta(days=24), d2)
        s = dt.datetime.combine(cur, dt.time(0, 0, 0))
        e = dt.datetime.combine(nxt, dt.time(23, 59, 59))
        x = None
        try:
            x = get_price(code, start_date=s, end_date=e, frequency='1m',
                          fields=['open', 'close'], skip_paused=True, fq=None)
        except Exception as ex:
            if _ERR[0] < 3:
                _ERR[0] += 1
                print('get_price 失败', code, cur, nxt, repr(ex))
        if x is not None and len(x) > 0:
            out.append(x.dropna())
        cur = nxt + dt.timedelta(days=1)
    if not out:
        return None
    return pd.concat(out)


def get_fut_1m(sym, d1, d2):
    """国债期货主力连续 1 分钟数据: 按『前一交易日的主力合约』分段取真实合约(避免未来函数), 并标记合约代码"""
    days = list(get_trade_days(start_date=d1, end_date=d2))
    dom = []
    for d in days:
        try:
            c = get_dominant_future(sym, d)
        except Exception:
            c = None
        dom.append(c if c else None)
    s = pd.Series(dom, index=days).shift(1).ffill().bfill()
    seg = (s != s.shift(1)).cumsum()
    out = []
    for sid, g in s.groupby(seg):
        code = g.iloc[0]
        x = get_1m(code, g.index[0], g.index[-1])
        if x is None or len(x) == 0:
            continue
        x = x.copy()
        x['contract'] = code
        out.append(x)
    return pd.concat(out)


def to_bars(x, L, with_contract=False):
    """1 分钟 -> L 分钟 bar (标签为 bar 结束时刻; 上午 9:30-11:30, 下午 13:00-15:00; 期货 15:00 后分钟丢弃)
    这样 ETF 与期货的 bar 口径完全一致, 避免不同平台 30m/60m 切分差异"""
    mm = np.asarray(x.index.hour * 60 + x.index.minute)
    keep = ((mm > 570) & (mm <= 690)) | ((mm > 780) & (mm <= 900))
    x = x[keep]
    mm = mm[keep]
    st = np.where(mm <= 690, 570, 780)
    k = np.ceil((mm - st) / float(L)).astype(int)
    end_mm = st + k * L
    key = pd.to_datetime(x.index.date) + pd.to_timedelta(end_mm, unit='m')
    g = x.groupby(key)
    out = pd.DataFrame({'open': g['open'].first(), 'close': g['close'].last()})
    cols = ['open', 'close']
    if with_contract:
        out['contract'] = g['contract'].last()
        cols.append('contract')
    return out[cols]


# ============================== 模型组件 ==============================
def kalman(y, x, init_n=INIT_N):
    """y = alpha + beta*x 的时变卡尔曼滤波 (状态随机游走). 返回新息 v 和对冲比率 beta"""
    n = len(y)
    X0 = np.column_stack([x[:init_n], np.ones(init_n)])
    th = np.linalg.lstsq(X0, y[:init_n], rcond=None)[0]       # [beta, alpha]
    res = y[:init_n] - X0.dot(th)
    Ve = max(res.var(), 1e-6)
    P = Ve * np.linalg.inv(X0.T.dot(X0)) * 10.0
    Vw = np.diag([KF_Q[0], KF_Q[1]])
    v = np.zeros(n)
    beta = np.zeros(n)
    for t in range(n):
        H = np.array([x[t], 1.0])
        P = P + Vw
        v[t] = y[t] - H.dot(th)
        PH = P.dot(H)
        S = H.dot(PH) + Ve
        K = PH / S
        th = th + K * v[t]
        P = P - np.outer(K, PH)
        beta[t] = th[0]
    return v, beta


def highfreq(v, levels=WAVE_LEVELS):
    """因果小波(à trous Haar)高频分量: 高频 = v - 第 levels 层近似, 不含未来信息"""
    a = v.copy()
    for j in range(levels):
        s = 2 ** j
        sh = np.concatenate([np.full(s, a[0]), a[:-s]])
        a = 0.5 * (a + sh)
    return v - a


def ddivf(w, train_win=400, refit=40, start=100):
    """数据驱动的新息波动率预测: 用 |w| 的多尺度滞后均值做滚动岭回归预测下一根 |w|, 再乘 sqrt(pi/2) 还原为标准差
    (研报未公开 DDIVF 细节, 此处为同类思想的因果实现; 预测 t 时刻仅用 t-1 及以前数据)"""
    n = len(w)
    a = np.abs(w)
    s = pd.Series(a)
    F = np.column_stack([np.ones(n), s.shift(1).values,
                         s.rolling(5).mean().shift(1).values,
                         s.rolling(20).mean().shift(1).values,
                         s.rolling(60).mean().shift(1).values])
    pred = np.full(n, np.nan)
    coef = None
    last = -10 ** 9
    for t in range(start, n):
        if t - last >= refit:
            lo = max(60, t - train_win)
            X = F[lo:t]
            y = a[lo:t]
            ok = ~np.isnan(X).any(axis=1)
            if ok.sum() >= 60:
                Xo, yo = X[ok], y[ok]
                coef = np.linalg.solve(Xo.T.dot(Xo) + 1e-8 * np.eye(5), Xo.T.dot(yo))
                last = t
        if coef is not None and not np.isnan(F[t]).any():
            pred[t] = max(F[t].dot(coef), 0.3 * F[t, 3])
    return pred * np.sqrt(np.pi / 2.0)


class GaussHMM(object):
    """对角协方差高斯 HMM (Baum-Welch, 缩放前后向), 仅用 numpy"""

    def __init__(self, k=2, n_iter=25, tol=1e-3):
        self.k, self.n_iter, self.tol = k, n_iter, tol

    def _logb(self, X):
        lb = np.zeros((len(X), self.k))
        for j in range(self.k):
            lb[:, j] = -0.5 * (np.log(2 * np.pi * self.var[j]).sum() +
                               (((X - self.mu[j]) ** 2) / self.var[j]).sum(axis=1))
        return lb

    def _mstep_emit(self, X, g):
        gs = g.sum(axis=0) + 1e-8
        self.mu = g.T.dot(X) / gs[:, None]
        self.var = np.array([(g[:, j:j + 1] * (X - self.mu[j]) ** 2).sum(axis=0) / gs[j]
                             for j in range(self.k)]) + 1e-3

    def _forward(self, B):
        n = B.shape[0]
        al = np.zeros_like(B)
        c = np.zeros(n)
        a = self.pi * B[0]
        c[0] = a.sum() + 1e-300
        al[0] = a / c[0]
        for t in range(1, n):
            a = al[t - 1].dot(self.A) * B[t]
            c[t] = a.sum() + 1e-300
            al[t] = a / c[t]
        return al, c

    def fit(self, X):
        n = X.shape[0]
        g = np.zeros((n, self.k))
        qs = np.percentile(X[:, 0], np.linspace(0, 100, self.k + 1)[1:-1])
        lab = np.searchsorted(qs, X[:, 0])
        g[np.arange(n), lab] = 1.0
        self._mstep_emit(X, g)
        self.A = np.full((self.k, self.k), 0.05 / max(self.k - 1, 1))
        np.fill_diagonal(self.A, 0.95)
        self.pi = np.full(self.k, 1.0 / self.k)
        prev = -np.inf
        for it in range(self.n_iter):
            lb = self._logb(X)
            m = lb.max(axis=1, keepdims=True)
            B = np.exp(lb - m) + 1e-300
            al, c = self._forward(B)
            ll = np.log(c).sum() + m.sum()
            be = np.ones_like(B)
            for t in range(n - 2, -1, -1):
                be[t] = self.A.dot(B[t + 1] * be[t + 1]) / c[t + 1]
            gam = al * be
            gam /= gam.sum(axis=1, keepdims=True)
            xi = self.A * al[:-1].T.dot(B[1:] * be[1:] / c[1:, None])
            self.A = xi / (xi.sum(axis=1, keepdims=True) + 1e-300)
            self.pi = gam[0] + 1e-8
            self.pi /= self.pi.sum()
            self._mstep_emit(X, gam)
            if abs(ll - prev) < self.tol * n * 1e-2:
                break
            prev = ll
        return self

    def filter_states(self, X):
        """因果滤波状态: 只用到 t 及以前的观测. 返回 0=平稳, 1=动荡(波动特征均值更大的状态)"""
        lb = self._logb(X)
        B = np.exp(lb - lb.max(axis=1, keepdims=True)) + 1e-300
        al, _ = self._forward(B)
        hi = int(np.argmax(self.mu[:, 0]))
        return (np.argmax(al, axis=1) == hi).astype(int)


# ============================== 参数选择 ==============================
def select_params(kind, state, w, m, sig, R, cost):
    """训练集网格遍历 (P0, P1): 返回净收益最大且 >0 的组合, 否则 None(不交易)
    kind='arb': 价差过小做多价差, 用对冲后价差收益; kind='rep': 价差过大多头替代, 用相对ETF的超额收益"""
    best = None
    for P0 in P_GRID:
        for P1 in P_GRID:
            thr = np.where(state == 0, P0, P1)
            if kind == 'arb':
                cond = w < m - thr * sig
            else:
                cond = w > m + thr * sig
            pos = cond.astype(float)
            pr = pos[:-1]
            chg = np.abs(np.diff(np.concatenate(([0.0], pr, [0.0])))).sum()
            pnl = (pr * R[1:]).sum() - chg * cost
            if best is None or pnl > best[0]:
                best = (pnl, P0, P1)
    if best is None or best[0] <= 0:
        return None
    return best


# ============================== 交易模拟 ==============================
def simulate(D, tgt, fee_on):
    """逐 bar 撮合. tgt[t]: t 时刻收盘产生的目标模式 (0=持ETF, 1=做多价差[ETF+空期货], 2=多头替代[期货多头]),
    在 t+1 根 bar 开盘价(含滑点)执行. 期货按 bar 盯市(换月缺口已用ETF对齐的调整价序列剔除)"""
    eo, ec, fo, fc, beta = D['eo'], D['ec'], D['fo_a'], D['fc_a'], D['beta']
    n = len(ec)
    es = ETF_SLIP_TICKS * TICK_ETF
    fs = FUT_SLIP_TICKS * D['tick_f']
    fr = ETF_FEE if fee_on else 0.0
    cash, sh, h, mode = INIT_CASH, 0, 0, -1
    slip, fee = 0.0, 0.0
    eq = np.zeros(n)
    trades = []
    cur = None
    for i in range(n):
        want = int(tgt[i - 1]) if i > 0 else 0
        if h != 0 and i > 0:
            cash += h * MULT * (fo[i] - fc[i - 1])
        if want != mode:
            eq_pre = cash + sh * eo[i]
            b = beta[i - 1] if i > 0 else 1.0
            sold = 0
            if h != 0 and (want == 0 or (want == 1 and h > 0) or (want == 2 and h < 0)):
                c = abs(h) * MULT * fs
                f = abs(h) * FUT_FEE
                cash -= c + f
                slip += c
                fee += f
                h = 0
            if want in (0, 1) and sh == 0:
                px = eo[i] + es
                ns = int(cash * ETF_RATIO / (px * (1 + fr)) / 100.0) * 100
                cash -= ns * px * (1 + fr)
                slip += ns * es
                fee += ns * px * fr
                sh = ns
            elif want == 2 and sh > 0:
                px = eo[i] - es
                cash += sh * px * (1 - fr)
                slip += sh * es
                fee += sh * px * fr
                sold = sh
                sh = 0
                hands = int(sold * b / MULT)
                if hands > 0:
                    c = hands * MULT * fs
                    f = hands * FUT_FEE
                    cash -= c + f
                    slip += c
                    fee += f
                    h = hands
            if want == 1 and sh > 0 and h == 0:
                hands = int(sh * b / MULT)
                if hands > 0:
                    c = hands * MULT * fs
                    f = hands * FUT_FEE
                    cash -= c + f
                    slip += c
                    fee += f
                    h = -hands
            eq_post = cash + sh * eo[i]
            prev = mode
            if cur is not None:
                pnl = eq_post - cur['eq0']
                ex = pnl - cur['sh_ref'] * (eo[i] - cur['eo0'])
                trades.append({'i0': cur['i0'], 'i1': i, 'mode': cur['mode'],
                               'm': ex if cur['mode'] == 2 else pnl})
                cur = None
            if want in (1, 2):
                cur = {'i0': i, 'mode': want, 'eo0': eo[i],
                       'eq0': eq_pre if prev in (0, -1) else eq_post,
                       'sh_ref': sold if want == 2 else sh}
            mode = want
        if h != 0:
            cash += h * MULT * (fc[i] - fo[i])
        eq[i] = cash + sh * ec[i]
    if cur is not None:
        pnl = eq[-1] - cur['eq0']
        ex = pnl - cur['sh_ref'] * (ec[-1] - cur['eo0'])
        trades.append({'i0': cur['i0'], 'i1': n - 1, 'mode': cur['mode'],
                       'm': ex if cur['mode'] == 2 else pnl})
    return eq, trades, slip, fee


# ============================== 绩效统计 ==============================
def daily_nav(eq, index):
    s = pd.Series(eq, index=index)
    d = s.groupby(s.index.date).last()
    return d / INIT_CASH


def perf(nav):
    n = len(nav)
    rets = nav.pct_change()
    rets.iloc[0] = nav.iloc[0] - 1.0
    ann = nav.iloc[-1] ** (252.0 / n) - 1.0
    dd = nav / nav.cummax() - 1.0
    mdd = dd.min()
    t_end = dd.idxmin()
    t_start = nav[:t_end].idxmax()
    sharpe = rets.mean() / rets.std() * np.sqrt(252.0) if rets.std() > 0 else np.nan
    calmar = ann / abs(mdd) if mdd != 0 else np.nan
    return {'ann': ann, 'mdd': mdd, 'sharpe': sharpe, 'calmar': calmar,
            't_start': t_start, 't_end': t_end}


def trade_stats(trades, index, L, ndays):
    if not trades:
        return {'plr': np.nan, 'win': np.nan, 'avg_n': 0.0, 'max_n': 0, 'hours': np.nan}
    m = np.array([t['m'] for t in trades])
    win = (m > 0).mean()
    w, l = m[m > 0], m[m <= 0]
    plr = w.mean() / abs(l.mean()) if len(w) and len(l) and l.mean() != 0 else np.nan
    days = pd.Series([index[t['i0']].date() for t in trades])
    cnt = days.value_counts()
    hours = np.mean([(t['i1'] - t['i0']) for t in trades]) * L / 60.0
    return {'plr': plr, 'win': win, 'avg_n': len(trades) / float(ndays),
            'max_n': int(cnt.max()), 'hours': hours}


def _p(x):
    return '{:.2f}%'.format(x * 100) if x == x else '/'


def _f(x):
    return '{:.2f}'.format(x) if x == x else '/'


def show_perf(title, navs, pfs, tss):
    """navs/pfs/tss: [ETF持有, 无ETF手续费, 有ETF手续费]"""
    names = ['国债ETF多头持有', '无国债ETF手续费', '有国债ETF手续费']
    rows = ['年化收益率', '最大回撤', '夏普比率', '卡玛比率', '盈亏比', '胜率', '最大回撤开始时间', '最大回撤结束时间']
    data = {}
    for k, nm in enumerate(names):
        p, t = pfs[k], tss[k]
        data[nm] = [_p(p['ann']), _p(p['mdd']), _f(p['sharpe']), _f(p['calmar']),
                    _f(t['plr']) if t else '/', _p(t['win']) if t else '/',
                    str(p['t_start']), str(p['t_end'])]
    df = pd.DataFrame(data, index=rows)[names]
    print('\n' + title)
    print(df.to_string())
    for k in (1, 2):
        print('  {} 年化收益增强: {:.2f}%'.format(names[k], (pfs[k]['ann'] - pfs[0]['ann']) * 100))


def show_trades(title, res, tss, ndays):
    slip, fee, pnl = res
    df = pd.DataFrame({'滑点支出(万)': ['{:.2f}'.format(slip / 1e4)],
                       '手续费支出(万)': ['{:.2f}'.format(fee / 1e4)],
                       '扣费后盈亏(万)': ['{:.2f}'.format(pnl / 1e4)],
                       '日均信号次数': ['{:.2f}'.format(tss['avg_n'])],
                       '每日最大信号次数': [tss['max_n']],
                       '平均信号时长(小时)': ['{:.2f}'.format(tss['hours'])]})
    print('\n' + title)
    print(df.to_string(index=False))


def plot_nav(title, navs, labels):
    if not PLOT:
        return
    try:
        import matplotlib.pyplot as plt
        plt.figure(figsize=(11, 4))
        for s, lb in zip(navs, labels):
            plt.plot(pd.to_datetime(s.index), s.values, label=lb)
        plt.title(title)
        plt.legend(loc='best')
        plt.grid(True, alpha=0.3)
        plt.show()
    except Exception:
        pass


# ============================== 单品种流程 ==============================
def run_case(case, e1m, f1m):
    etf, sym, L = case['etf'], case['fut'], case['L']
    try:
        info = get_security_info(etf)
        print('\n' + '=' * 78)
        print('{}  {}  | 期货:{} 主力 | {}分钟线'.format(etf, info.display_name, sym, L))
    except Exception:
        print('\n' + '=' * 78)
        print('{} | 期货:{} 主力 | {}分钟线'.format(etf, sym, L))

    e = to_bars(e1m, L)
    f = to_bars(f1m, L, with_contract=True)
    e.columns = ['eo', 'ec']
    f.columns = ['fo', 'fc', 'contract']
    df = e.join(f, how='inner').dropna()
    n = len(df)
    print('对齐后 bar 数: {}, 区间 {} ~ {}'.format(n, df.index[0], df.index[-1]))
    eo, ec = df['eo'].values, df['ec'].values
    fo, fc = df['fo'].values, df['fc'].values
    ctr = df['contract'].values

    # ---- 换月处理: 用 ETF 同期涨跌对齐换月跳空, 得到连续的期货调整价(仅用于模型与盯市) ----
    off = np.zeros(n)
    cum = 0.0
    roll = np.zeros(n, dtype=bool)
    for i in range(1, n):
        if ctr[i] != ctr[i - 1]:
            roll[i] = True
            cum += (fo[i] - fc[i - 1]) - (eo[i] - ec[i - 1])
        off[i] = cum
    fo_a, fc_a = fo - off, fc - off

    # ---- 模型序列 (全部因果) ----
    v, beta = kalman(ec, fc_a)
    w = highfreq(v)
    m = pd.Series(w).rolling(MEAN_WIN).mean().shift(1).values
    sig = ddivf(w)
    sig[:VALID_FROM] = np.nan
    re = np.r_[0.0, np.diff(np.log(ec))]
    rf = np.r_[0.0, np.diff(fc_a) / fc[:-1]]
    feats = np.column_stack([
        np.log(pd.Series(re).rolling(20).std().values + 1e-6),
        np.log(pd.Series(rf).rolling(20).std().values + 1e-6),
        np.log(pd.Series(np.abs(w)).rolling(20).mean().values + 1e-6)])
    feats[:VALID_FROM] = np.nan

    # 用于训练的逐 bar 收益(占ETF价格比例): 持有 t-1 的仓位在第 t 根 bar 获得 R[t]
    dE = np.r_[0.0, np.diff(ec)]
    dF = np.r_[0.0, np.diff(fc_a)]
    b_prev = np.r_[beta[0], beta[:-1]]
    e_prev = np.r_[ec[0], ec[:-1]]
    R_rep = (b_prev * dF - dE) / e_prev      # 期货多头 相对 ETF 的超额
    R_arb = -R_rep                           # 持ETF空期货 的对冲价差收益
    R_rep = np.nan_to_num(R_rep)
    R_arb = np.nan_to_num(R_arb)

    tick_f = TICK_FUT[sym]
    fut_c = np.mean(beta) * (FUT_SLIP_TICKS * tick_f + FUT_FEE / MULT) / np.mean(ec)
    etf_c0 = ETF_SLIP_TICKS * TICK_ETF / np.mean(ec)
    cost_arb = fut_c
    cost_rep = {False: etf_c0 + fut_c, True: etf_c0 + ETF_FEE + fut_c}

    # ---- 日期边界 ----
    dates = df.index.date
    first_pos = pd.Series(np.arange(n)).groupby(dates).first()
    day_list = list(first_pos.index)
    day_start = list(first_pos.values) + [n]
    nd = len(day_list)
    d_begin = max(bisect.bisect_left(day_list, START), 30)
    if d_begin >= nd:
        print('数据不足, 跳过')
        return
    sim_start = day_start[d_begin]

    tgt_rep = {False: np.zeros(n), True: np.zeros(n)}
    tgt_ovl = {False: np.zeros(n), True: np.zeros(n)}
    log = []
    for d0 in range(d_begin, nd, STEP_DAYS):
        d1 = min(d0 + STEP_DAYS, nd)
        a = day_start[max(0, d0 - TRAIN_DAYS)]
        b = day_start[d0]
        c = day_start[d1]
        X = feats[a:c]
        ok = ~np.isnan(X).any(axis=1)
        okt = ok[:b - a]
        if okt.sum() < 150 or (c - b) <= 0:
            continue
        mu = X[:b - a][okt].mean(axis=0)
        sd = X[:b - a][okt].std(axis=0) + 1e-8
        st = np.zeros(c - a, dtype=int)
        try:
            hmm = GaussHMM().fit((X[:b - a][okt] - mu) / sd)
            st[ok] = hmm.filter_states((X[ok] - mu) / sd)
        except Exception as ex:
            print('HMM 失败', repr(ex))
        s_tr, s_te = st[:b - a], st[b - a:]
        sl_tr, sl_te = slice(a, b), slice(b, c)
        pa = select_params('arb', s_tr, w[sl_tr], m[sl_tr], sig[sl_tr], R_arb[sl_tr], cost_arb)
        cond_arb = np.zeros(c - b, dtype=bool)
        if pa is not None:
            thr = np.where(s_te == 0, pa[1], pa[2])
            cond_arb = w[sl_te] < m[sl_te] - thr * sig[sl_te]
        rec = [str(day_list[d0]), '-' if pa is None else '({},{})'.format(pa[1], pa[2])]
        for fee_on in (False, True):
            pr = select_params('rep', s_tr, w[sl_tr], m[sl_tr], sig[sl_tr], R_rep[sl_tr], cost_rep[fee_on])
            cond_rep = np.zeros(c - b, dtype=bool)
            if pr is not None:
                thr = np.where(s_te == 0, pr[1], pr[2])
                cond_rep = w[sl_te] > m[sl_te] + thr * sig[sl_te]
            rec.append('-' if pr is None else '({},{})'.format(pr[1], pr[2]))
            tgt_rep[fee_on][b:c] = 2 * cond_rep
            tgt_ovl[fee_on][b:c] = np.where(cond_rep, 2, np.where(cond_arb, 1, 0))
        log.append(rec)

    # 换月前后 bar 不开新仓/强制回到ETF
    for r in np.where(roll)[0]:
        for T in (tgt_rep, tgt_ovl):
            for k in (False, True):
                T[k][max(r - 1, 0):r + 3] = 0

    print('\n滚动参数(起始日, 做多价差P0/P1, 多头替代P0/P1[无ETF费], 多头替代P0/P1[有ETF费]); "-"=训练集无超额, 空仓观察:')
    print(pd.DataFrame(log, columns=['起始日', '做多价差', '替代(无费)', '替代(有费)']).to_string(index=False))

    # ---- 回测 ----
    S = slice(sim_start, n)
    D = {'eo': eo[S], 'ec': ec[S], 'fo_a': fo_a[S], 'fc_a': fc_a[S], 'beta': beta[S], 'tick_f': tick_f}
    idx = df.index[S]
    eq_b, _, _, _ = simulate(D, np.zeros(n - sim_start), False)
    nav_b = daily_nav(eq_b, idx)
    pf_b = perf(nav_b)
    ndays = len(nav_b)
    print('\n回测区间: {} ~ {} (共{}个交易日), 初始资金 {:.0f} 万'.format(
        idx[0].date(), idx[-1].date(), ndays, INIT_CASH / 1e4))

    for name, TG in (('多头替代增强策略', tgt_rep), ('统计套利+多头替代 叠加策略', tgt_ovl)):
        navs, pfs, tss, extra = [nav_b], [pf_b], [None], {}
        for fee_on in (False, True):
            eq, trades, slip, fee = simulate(D, TG[fee_on][S], fee_on)
            nv = daily_nav(eq, idx)
            navs.append(nv)
            pfs.append(perf(nv))
            ts = trade_stats(trades, idx, L, ndays)
            tss.append(ts)
            extra[fee_on] = (slip, fee, eq[-1] - INIT_CASH, ts)
        show_perf('[{}] {} 净值表现对比'.format(etf, name), navs, pfs, tss)
        slip, fee, pnl, ts = extra[True]
        show_trades('[{}] {} 收取ETF手续费情况下的交易统计'.format(etf, name), (slip, fee, pnl), ts, ndays)
        ex0 = navs[1] - navs[0] + 1.0
        ex1 = navs[2] - navs[0] + 1.0
        plot_nav('{} {}'.format(etf, name), [navs[1], navs[2], navs[0], ex1],
                 ['no-ETF-fee', 'with-ETF-fee', 'buy&hold', 'excess(with fee)'])


def main():
    print('回测区间 {} ~ {}, 预热起点 {}'.format(START, END, WARM_START))
    print('说明: 研报正文把 511130/511090 的基金名称写反; 实际 511130=博时上证30年国债ETF, 511090=鹏扬中债30年国债ETF, 以聚宽返回名称为准')
    f_cache, e_cache = {}, {}
    for case in CASES:
        sym, etf = case['fut'], case['etf']
        if sym not in f_cache:
            f_cache[sym] = get_fut_1m(sym, WARM_START, END)
        if etf not in e_cache:
            e_cache[etf] = get_1m(etf, WARM_START, END)
        if e_cache[etf] is None or f_cache[sym] is None:
            print(etf, '无数据, 跳过')
            continue
        run_case(case, e_cache[etf], f_cache[sym])


main()
```
