```python
# -*- coding: utf-8 -*-
"""
湘财证券《商品期货的多因子应用之因子筛选》(CTA系列专题三) 复现 —— 聚宽研究环境版
------------------------------------------------------------------------------
环境: Python 3.6.7 / numpy 1.19.5 / pandas 0.23.4 / scipy 1.2.0 (聚宽研究)
数据: 全部来自聚宽 (get_all_securities / get_price / finance.FUT_WAREHOUSE_RECEIPT)
输出: 不写磁盘, 数据取来即用, 结果直接 print (+ plt.show 画五组分层净值)

流程(与研报一致):
  1) 构建 36 个因子 (动量12 / 期限3 / 量价5 / 波动率8 / 库存3 / 持仓5)
  2) 月频、周频两种调仓频率, 截面因子值 -> 下一调仓期收益
  3) 计算 IR / Rank IR / IC / Rank IC / 多空收益 (对应研报表7、表8)
  4) 五组分层净值 (对应研报图1~图26), 筛出有效因子

关键口径(研报未写明处, 均可在下方参数区修改):
  - 主力合约: 持仓量最大, 且只允许向更远月切换(避免来回跳); 收益用"前一日持有的主力合约"
    当日涨跌计算, 因此没有换月跳空, 也没有未来函数
  - 近月/远月合约: 当日持仓量最大的两个合约, 按到期日先后分为近月(P1)与远月(P2)
  - 展期收益率 = (ln P1 - ln P2) * 365 / (N2 - N1)   (研报公式中 ln(P1-P2) 应为笔误)
  - 持仓类因子用品种所有合约合计持仓量/持仓金额(持仓量*收盘价)
  - 仓单 = 品种当日所有仓库仓单数量之和 (finance.FUT_WAREHOUSE_RECEIPT)
  - group1 = 因子值最高组, group5 = 最低组
  - 多空收益: 方向取 sign(IC均值) (IC为负的因子反向操作), 多空 = group1 - group5, 年化=期均收益*年期数
  - 日内波动趋势因子里的 RS/GK/PK/YZ 取"单日"估计值; YZ 窗口版按标准 Yang-Zhang 公式
  - 回测样本只有 2025-01 ~ 2026-10, 约 21 个月/90 周, 统计结果与研报(2015~2025.2)会有差异
"""
import re
import datetime
import warnings
from collections import OrderedDict

import gc
import numpy as np
import pandas as pd

warnings.filterwarnings('ignore')

from jqdata import finance
try:
    query
except NameError:
    from jqdata import *

# ============================ 参数区 ============================
START_DATE = '2025-01-01'          # 因子测试区间起点
END_DATE = '2026-10-20'            # 因子测试区间终点 (超过今天自动截断)
WARMUP_START = '2024-07-01'        # 因子预热数据起点 (滚动窗口需要历史)
N_GROUP = 5                        # 分层组数
MIN_XS = 10                        # 单期截面最少品种数
IR_THRESHOLD = 0.10                # 标记"有效因子": |IR|与|RankIR| 均 >= 该值
PLOT = True                        # 是否画五组分层净值图
PLOT_TOP = 6                       # 每个频率画 |IR| 最大的前几个因子
WH_CHUNK_DAYS = 45                 # 仓单查询时间块 (单次最多返回4000行, 超限自动二分)

FREQ_PARAMS = OrderedDict([
    ('月频', dict(rule='M', N=20, fast=5, slow=20, ppy=12)),
    ('周频', dict(rule='W', N=10, fast=3, slow=10, ppy=52)),
])
N_IDIO = 20                        # 特质波动率回归窗口 (日)

FIELDS = ['open', 'close', 'high', 'low', 'volume', 'money', 'open_interest']

CATS = OrderedDict([
    ('动量因子', ['主力涨幅', '上涨次数占比', '涨幅占比', '均线偏离比值', '快慢均线偏离率',
              '隔夜趋势因子', '日内动量因子', '日内累计振幅因子', '日内RS波动趋势因子',
              '日内GK波动趋势因子', '日内PK波动趋势因子', '日内YZ波动趋势因子']),
    ('期限因子', ['展期收益率', '基差因子', '基差动量']),
    ('量价因子', ['偏度因子', '流动性因子', '非流动性因子', '均价突破因子', '标准化均价突破因子']),
    ('波动率因子', ['变异系数因子', '变异系数因子2', '特质波动率', 'RS波动率因子', 'GK波动率因子',
               'PK波动率因子', 'YZ波动率因子', '波动率因子']),
    ('库存因子', ['仓单水平', '仓单变化比值', '仓单涨幅']),
    ('持仓因子', ['持仓量变化比值', '持仓量涨幅', '持仓金额变化比值', '持仓金额涨幅', '对冲压力因子']),
])
LN2 = np.log(2.0)


# ============================ 数据获取 ============================
def list_contracts(win_start, win_end):
    """商品期货所有具体合约(剔除金融期货CCFX, 剔除9999/8888连续合约)"""
    sec = get_all_securities(['futures'])
    idx = pd.Series(sec.index, index=sec.index)
    ok = idx.str.match(r'^[A-Za-z]{1,2}\d{3,4}\.(XDCE|XZCE|XSGE|XINE|GFEX)$')
    ok = ok & (~idx.str.contains(r'(9999|8888)\.'))
    sec = sec[ok.values].copy()
    sec['start_date'] = pd.to_datetime(sec['start_date'])
    sec['end_date'] = pd.to_datetime(sec['end_date'])
    sec = sec[(sec['end_date'] >= win_start) & (sec['start_date'] <= win_end)]
    out = pd.DataFrame({'code': list(sec.index), 'end': sec['end_date'].values})
    out['variety'] = out['code'].str.extract(r'^([A-Za-z]+)', expand=False).str.upper()
    return out


def fetch_prices(codes, start, end, chunk=20, verbose=True):
    frames = []
    for i in range(0, len(codes), chunk):
        part = codes[i:i + chunk]
        try:
            df = get_price(part, start_date=start, end_date=end, frequency='daily',
                           fields=FIELDS, skip_paused=False, panel=False)
        except Exception as e:
            print('  批量取价失败, 逐个重试:', str(e)[:60])
            df = None
            sub = []
            for c in part:
                try:
                    one = get_price(c, start_date=start, end_date=end, frequency='daily',
                                    fields=FIELDS, skip_paused=False)
                    one = one.copy()
                    one['code'] = c
                    one['time'] = one.index
                    sub.append(one)
                except Exception:
                    pass
            if sub:
                df = pd.concat(sub, ignore_index=True)
        if df is not None and len(df) > 0:
            frames.append(df)
        if verbose:
            print('  行情进度 %d/%d' % (min(i + chunk, len(codes)), len(codes)))
    if not frames:
        return None
    raw = pd.concat(frames, ignore_index=True)
    raw['time'] = pd.to_datetime(raw['time'])
    raw = raw.drop_duplicates(['time', 'code'])
    return raw


def _wh_fetch(code, d0, d1, out):
    T = finance.FUT_WAREHOUSE_RECEIPT
    q = query(T.day, T.underlying_code, T.warehouse_name, T.warehouse_receipt_number).filter(
        T.underlying_code == code, T.day >= d0, T.day <= d1)
    df = finance.run_query(q)
    if df is None or len(df) == 0:
        return
    if len(df) >= 3990 and (d1 - d0).days >= 1:       # 触到4000行上限 -> 二分
        mid = d0 + datetime.timedelta(days=(d1 - d0).days // 2)
        _wh_fetch(code, d0, mid, out)
        _wh_fetch(code, mid + datetime.timedelta(days=1), d1, out)
    else:
        out.append(df)


def fetch_warehouse(variety, d0, d1):
    """某品种每日仓单合计 (Series, index=日期)"""
    out = []
    cur = d0
    while cur <= d1:
        nxt = min(cur + datetime.timedelta(days=WH_CHUNK_DAYS - 1), d1)
        try:
            _wh_fetch(variety, cur, nxt, out)
        except Exception as e:
            print('  仓单查询失败', variety, cur, str(e)[:50])
        cur = nxt + datetime.timedelta(days=1)
    if not out:
        return None
    df = pd.concat(out, ignore_index=True)
    df['day'] = pd.to_datetime(df['day'])
    df['warehouse_receipt_number'] = df['warehouse_receipt_number'].astype(float)
    df = df.drop_duplicates(['day', 'warehouse_name'])
    return df.groupby('day')['warehouse_receipt_number'].sum()


# ============================ 品种级序列构建 ============================
def _pick(M, idx, lag=0):
    """out[i] = M[i-lag, idx[i]]"""
    T = len(idx)
    out = np.full(T, np.nan)
    rows = np.arange(T) - lag
    ok = (idx >= 0) & (rows >= 0)
    out[ok] = M[rows[ok], idx[ok]]
    return out


def build_variety(cols, end_ts, W, dates):
    C = W['close'][cols].values
    O = W['open'][cols].values
    H = W['high'][cols].values
    L = W['low'][cols].values
    V = W['volume'][cols].values
    M = W['money'][cols].values
    OI = W['open_interest'][cols].values
    T = len(dates)
    ends = pd.DatetimeIndex([end_ts[c] for c in cols]).values
    dts = dates.values

    main = np.full(T, -1, dtype=int)
    near = np.full(T, -1, dtype=int)
    far = np.full(T, -1, dtype=int)
    cur = -1
    with np.errstate(invalid='ignore'):
        for i in range(T):
            valid = (V[i] > 0) & (OI[i] > 0) & (ends >= dts[i]) & (~np.isnan(C[i]))
            if not valid.any():
                cur = -1
                continue
            oi_v = np.where(valid, OI[i], -1.0)
            cand = int(np.argmax(oi_v))
            if cur == -1 or (not valid[cur]) or (cand != cur and ends[cand] > ends[cur]):
                cur = cand
            main[i] = cur
            if valid.sum() >= 2:
                a, b = np.argsort(-oi_v)[:2]
                if ends[a] > ends[b]:
                    a, b = b, a
                near[i], far[i] = a, b

    held = np.concatenate([[-1], main[:-1]])          # 前一日决定, 当日持有
    d = pd.DataFrame(index=dates)
    d['o'] = _pick(O, held)
    d['h'] = _pick(H, held)
    d['l'] = _pick(L, held)
    d['c'] = _pick(C, held)
    d['pc'] = _pick(C, held, 1)
    d['v'] = _pick(V, held)
    d['m'] = _pick(M, held)
    d['ret'] = d['c'] / d['pc'] - 1
    d.loc[d['ret'].abs() > 0.5, 'ret'] = np.nan
    first = d['ret'].first_valid_index()
    if first is None:
        return None
    r0 = d['ret'].copy()
    r0.loc[first:] = r0.loc[first:].fillna(0.0)
    I = (1 + r0).cumprod()                            # 连续(换月无跳空)价格指数

    with np.errstate(invalid='ignore'):
        tot_oi = pd.Series(np.nansum(OI, axis=1), index=dates).replace(0, np.nan)
        tot_amt = pd.Series(np.nansum(OI * C, axis=1), index=dates).replace(0, np.nan)
        tot_vol = pd.Series(np.nansum(V, axis=1), index=dates).replace(0, np.nan)

    nr = near >= 0
    N1 = np.full(T, np.nan)
    N2 = np.full(T, np.nan)
    N1[nr] = (ends[near[nr]] - dts[nr]) / np.timedelta64(1, 'D')
    N2[nr] = (ends[far[nr]] - dts[nr]) / np.timedelta64(1, 'D')
    term = pd.DataFrame({'P1': _pick(C, near), 'P2': _pick(C, far), 'N1': N1, 'N2': N2}, index=dates)

    return dict(d=d, I=I, tot_oi=tot_oi, tot_amt=tot_amt, tot_vol=tot_vol,
                term=term, C=C, near=near, far=far)


def basis_momentum(vd, N):
    cur = lambda idx: _pick(vd['C'], idx, 0) / _pick(vd['C'], idx, N) - 1
    return pd.Series(cur(vd['near']) - cur(vd['far']), index=vd['d'].index)


# ============================ 因子计算 ============================
def compute_factors(vd, P, wh):
    N, fast, slow = P['N'], P['fast'], P['slow']
    mp = max(3, int(N * 0.7))
    d, I = vd['d'], vd['I']
    ret = d['ret']
    F = OrderedDict()

    # ---- 动量因子 ----
    F['主力涨幅'] = I / I.shift(N - 1) - 1
    up = (ret > 0).astype(float).where(ret.notna())
    F['上涨次数占比'] = up.rolling(N, min_periods=mp).mean()
    F['涨幅占比'] = ret.clip(lower=0).rolling(N, min_periods=mp).sum() / ret.abs().rolling(N, min_periods=mp).sum()
    F['均线偏离比值'] = -I.rolling(N, min_periods=mp).mean() / I
    mf, ms = I.rolling(fast).mean(), I.rolling(slow).mean()
    F['快慢均线偏离率'] = (mf - ms) / ms
    ovn = d['o'] / d['pc'] - 1
    intr = d['c'] / d['o'] - 1
    F['隔夜趋势因子'] = ovn.rolling(N, min_periods=mp).mean()
    F['日内动量因子'] = intr.rolling(N, min_periods=mp).mean()
    sg = np.sign(d['c'] - d['o'])
    amp = (2 * (d['h'] - d['l']) * sg - (d['c'] - d['o'])) / d['c']
    F['日内累计振幅因子'] = amp.rolling(N, min_periods=mp).mean()

    lhl = np.log(d['h'] / d['l'])
    lco = np.log(d['c'] / d['o'])
    rs_d = np.log(d['h'] / d['c']) * np.log(d['h'] / d['o']) + np.log(d['l'] / d['c']) * np.log(d['l'] / d['o'])
    gk_d = 0.5 * lhl ** 2 - (2 * LN2 - 1) * lco ** 2
    pk_d = lhl ** 2 / (4 * LN2)
    k = 0.34 / (1.34 + (N + 1.0) / (N - 1.0))
    o_d = np.log(d['o'] / d['pc'])
    yz_d = o_d ** 2 + k * lco ** 2 + (1 - k) * rs_d
    for nm, comp in [('RS', rs_d), ('GK', gk_d), ('PK', pk_d), ('YZ', yz_d)]:
        F['日内%s波动趋势因子' % nm] = (sg * np.sqrt(comp.clip(lower=0))).rolling(N, min_periods=mp).mean()

    # ---- 期限因子 ----
    t = vd['term']
    dn = (t['N2'] - t['N1']).where((t['N2'] - t['N1']) > 0)
    F['展期收益率'] = (np.log(t['P1']) - np.log(t['P2'])) * 365.0 / dn
    F['基差因子'] = (t['P1'] / t['P2'] - 1) * 365.0 / dn
    F['基差动量'] = basis_momentum(vd, N)

    # ---- 量价因子 ----
    F['偏度因子'] = ret.rolling(N, min_periods=mp).skew()
    illiq = ret.abs() / d['m'].replace(0, np.nan)
    F['非流动性因子'] = illiq.rolling(N, min_periods=mp).mean()
    liq = d['v'] / ret.abs().replace(0, np.nan)
    F['流动性因子'] = liq.rolling(N, min_periods=mp).mean()
    ma = I.rolling(N, min_periods=mp).mean()
    sd = ret.rolling(N, min_periods=mp).std()
    F['均价突破因子'] = I - ma
    F['标准化均价突破因子'] = (I - ma) / (sd * I.shift(N - 1))

    # ---- 波动率因子 ----
    mu = ret.rolling(N, min_periods=mp).mean()
    F['变异系数因子'] = sd ** 2 / mu
    F['变异系数因子2'] = sd / mu
    F['特质波动率'] = pd.Series(np.nan, index=d.index)       # 后面统一回填
    F['RS波动率因子'] = np.sqrt(rs_d.rolling(N, min_periods=mp).mean().clip(lower=0))
    F['GK波动率因子'] = np.sqrt(gk_d.rolling(N, min_periods=mp).mean().clip(lower=0))
    F['PK波动率因子'] = np.sqrt(pk_d.rolling(N, min_periods=mp).mean().clip(lower=0))
    var_o = o_d.rolling(N, min_periods=mp).var()
    var_c = lco.rolling(N, min_periods=mp).var()
    F['YZ波动率因子'] = np.sqrt((var_o + k * var_c + (1 - k) * rs_d.rolling(N, min_periods=mp).mean()).clip(lower=0))
    F['波动率因子'] = sd

    # ---- 库存因子 ----
    if wh is None:
        wh = pd.Series(np.nan, index=d.index)
    F['仓单水平'] = wh / wh.shift(1).rolling(N, min_periods=mp).mean()
    F['仓单变化比值'] = wh / wh.shift(N)
    F['仓单涨幅'] = wh / wh.shift(N) - 1

    # ---- 持仓因子 ----
    oi, amt, vol = vd['tot_oi'], vd['tot_amt'], vd['tot_vol']
    F['持仓量变化比值'] = oi / oi.shift(N)
    F['持仓量涨幅'] = oi / oi.shift(N) - 1
    F['持仓金额变化比值'] = amt / amt.shift(N)
    F['持仓金额涨幅'] = amt / amt.shift(N) - 1
    F['对冲压力因子'] = (oi - oi.shift(N)) / vol.rolling(N, min_periods=mp).sum()

    out = pd.DataFrame(F)
    return out.replace([np.inf, -np.inf], np.nan)


def ls_series(Fp, RET, q=0.3):
    """用上一日因子排序, 构造次日 多(前30%) - 空(后30%) 因子收益"""
    rk = Fp.shift(1).rank(axis=1, pct=True)
    lo = RET.where(rk >= 1 - q).mean(axis=1)
    sh = RET.where(rk <= q).mean(axis=1)
    return (lo - sh).fillna(0.0)


def idio_vol(RET, MKT, MOM, BAS, N):
    """对 市场/动量/基差 因子收益滚动回归, 残差标准差"""
    T = len(RET)
    X = np.column_stack([MKT.values, MOM.values, BAS.values, np.ones(T)])
    out = pd.DataFrame(np.nan, index=RET.index, columns=RET.columns)
    need = max(8, int(N * 0.7))
    for v in RET.columns:
        y = RET[v].values
        res = np.full(T, np.nan)
        for i in range(N - 1, T):
            yy = y[i - N + 1:i + 1]
            xx = X[i - N + 1:i + 1]
            ok = (~np.isnan(yy)) & (~np.isnan(xx).any(axis=1))
            if ok.sum() < need:
                continue
            beta = np.linalg.lstsq(xx[ok], yy[ok], rcond=None)[0]
            e = yy[ok] - xx[ok].dot(beta)
            res[i] = np.std(e, ddof=4)
        out[v] = res
    return out


def winsor_cs(F, k=3.0):
    med = F.median(axis=1)
    mad = F.sub(med, axis=0).abs().median(axis=1) * 1.4826
    mad = mad.replace(0, np.nan)
    return F.clip(lower=med - k * mad, upper=med + k * mad, axis=0)


# ============================ 因子评价 ============================
def rebalance_dates(idx, rule, start, end):
    idx = idx[(idx >= start) & (idx <= end)]
    s = pd.Series(idx, index=idx)
    g = s.groupby(idx.to_period(rule)).max()
    rd = pd.DatetimeIndex(sorted(g.values))
    last_per = rd[-1].to_period(rule)
    if (last_per.end_time.normalize() - rd[-1]).days > 3:     # 最后一期未走完 -> 丢弃
        rd = rd[:-1]
    return rd


def evaluate(Fp, FWD, rd):
    Fr = Fp.reindex(rd)
    ics, rics, grps, dts = [], [], [], []
    for t in rd[:-1]:
        f, r = Fr.loc[t], FWD.loc[t]
        m = f.notna() & r.notna()
        if m.sum() < MIN_XS:
            continue
        f, r = f[m], r[m]
        ic = np.corrcoef(f.values, r.values)[0, 1]
        ric = np.corrcoef(f.rank().values, r.rank().values)[0, 1]
        g = pd.qcut(f.rank(method='first', ascending=False), N_GROUP, labels=False)
        gr = r.groupby(g.values).mean().reindex(range(N_GROUP))
        ics.append(ic)
        rics.append(ric)
        grps.append(gr.values)
        dts.append(t)
    if not dts:
        return None
    ics, rics = pd.Series(ics, index=dts), pd.Series(rics, index=dts)
    G = pd.DataFrame(grps, index=dts, columns=['group%d' % (i + 1) for i in range(N_GROUP)])
    return ics, rics, G


def summarize(ics, rics, G, ppy):
    IC, RIC = ics.mean(), rics.mean()
    IR = IC / ics.std() if ics.std() > 0 else np.nan
    RIR = RIC / rics.std() if rics.std() > 0 else np.nan
    dirn = 1.0 if IC >= 0 else -1.0
    ls = dirn * (G['group1'] - G['group%d' % N_GROUP])
    return dict(IR=IR, RankIR=RIR, IC=IC, RankIC=RIC, LS=ls.mean() * ppy, dir=dirn, n=len(ics))

def get_calendar(start, end):
    """交易日历: 优先 get_trade_days, 取不到则用螺纹钢主力连续的交易日, 最后退回工作日"""
    s, e = start.strftime('%Y-%m-%d'), end.strftime('%Y-%m-%d')
    fn = globals().get('get_trade_days')
    if fn is None:
        try:
            from jqdata import get_trade_days as fn
        except Exception:
            fn = None
    if fn is not None:
        try:
            return pd.DatetimeIndex(pd.to_datetime(list(fn(start_date=s, end_date=e))))
        except Exception:
            pass
    try:
        df = get_price('RB9999.XSGE', start_date=s, end_date=e, frequency='daily', fields=['close'], skip_paused=False)
        return pd.DatetimeIndex(pd.to_datetime(df.index))
    except Exception:
        return pd.bdate_range(start, end)

# ============================ 主流程 ============================
def main():
    pd.set_option('display.width', 220)
    pd.set_option('display.max_rows', 200)
    pd.set_option('display.unicode.east_asian_width', True)

    win_start = pd.Timestamp(WARMUP_START)
    test_start = pd.Timestamp(START_DATE)
    today = pd.Timestamp(datetime.date.today())
    test_end = min(pd.Timestamp(END_DATE), today)
    print('测试区间: %s ~ %s  (END_DATE 超过今天会自动截断)' % (test_start.date(), test_end.date()))

    # 1) 合约列表与行情
    ctr = list_contracts(win_start, test_end)
    print('商品期货合约数: %d, 品种数: %d' % (len(ctr), ctr['variety'].nunique()))
    # 交易日历 (只保留日期, 很小); 按品种逐个取数->构建->释放, 避免一次性载入全部合约导致内存爆掉
    dates = get_calendar(win_start, test_end)
    end_ts = dict(zip(ctr['code'], ctr['end']))
    VD = OrderedDict()
    WH = {}
    miss = []
    varieties = sorted(ctr['variety'].unique())
    for k, v in enumerate(varieties):
        codes = list(ctr.loc[ctr['variety'] == v, 'code'])
        raw = fetch_prices(codes, win_start.strftime('%Y-%m-%d'), test_end.strftime('%Y-%m-%d'), chunk=30, verbose=False)
        if raw is None or len(raw) == 0:
            continue
        W = {f: raw.pivot(index='time', columns='code', values=f).reindex(dates) for f in FIELDS}
        del raw
        cols = [c for c in codes if c in W['close'].columns]
        res = build_variety(cols, end_ts, W, dates) if cols else None
        del W
        if res is None or res['d']['c'].notna().sum() < 60:
            gc.collect()
            continue
        # 只保留后续要用的列, 压缩为 float32
        res['d'] = res['d'].astype('float32')
        res['C'] = res['C'].astype('float32')
        VD[v] = res
        s_wh = fetch_warehouse(v, win_start.date(), test_end.date())
        if s_wh is None or len(s_wh) == 0:
            miss.append(v)
        else:
            WH[v] = s_wh.reindex(dates).ffill(limit=5).astype('float32')
        gc.collect()
        print('  品种进度 %d/%d  %s' % (k + 1, len(varieties), v))
    print('可用品种数: %d' % len(VD))
    print('品种:', ' '.join(VD.keys()))
    print('仓单数据缺失品种(库存因子对其为空):', miss if miss else '无')

    RET = pd.DataFrame(OrderedDict([(v, VD[v]['d']['ret']) for v in VD]))
    IDX = pd.DataFrame(OrderedDict([(v, VD[v]['I']) for v in VD]))
    TRD = pd.DataFrame(OrderedDict([(v, VD[v]['d']['c'].notna()) for v in VD]))

    # 4) 各频率因子面板
    PANELS = OrderedDict()
    for fname, P in FREQ_PARAMS.items():
        per = OrderedDict((v, compute_factors(VD[v], P, WH.get(v))) for v in VD)
        pan = OrderedDict()
        for nm in [x for names in CATS.values() for x in names]:
            pan[nm] = pd.DataFrame(OrderedDict([(v, per[v][nm]) for v in VD]))
        PANELS[fname] = pan

    # 5) 特质波动率: 对 市场/动量/基差 因子收益回归 (因子收益用首个频率的 主力涨幅 / 展期收益率 排序构造)
    first_pan = PANELS[list(FREQ_PARAMS.keys())[0]]
    MKT = RET.mean(axis=1).fillna(0.0)
    MOM = ls_series(first_pan['主力涨幅'], RET)
    BAS = ls_series(first_pan['展期收益率'], RET)
    print('计算特质波动率 (滚动回归) ...')
    IV = idio_vol(RET, MKT, MOM, BAS, N_IDIO)
    for fname in PANELS:
        PANELS[fname]['特质波动率'] = IV

    # 6) 因子评价
    plt = None
    if PLOT:
        try:
            import matplotlib.pyplot as plt
            plt.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei', 'WenQuanYi Micro Hei', 'DejaVu Sans']
            plt.rcParams['axes.unicode_minus'] = False
        except Exception:
            plt = None

    all_res = OrderedDict()
    for fname, P in FREQ_PARAMS.items():
        rd = rebalance_dates(dates, P['rule'], test_start, test_end)
        Ir = IDX.reindex(rd)
        FWD = Ir.shift(-1) / Ir - 1
        okm = TRD.reindex(rd)
        FWD = FWD.where(okm & okm.shift(-1).fillna(False))
        print('\n' + '=' * 100)
        print('【%s调仓】 窗口N=%d, 调仓期数=%d (%s ~ %s)' % (fname, P['N'], len(rd) - 1, rd[0].date(), rd[-1].date()))
        rows, details = [], OrderedDict()
        for cat, names in CATS.items():
            for nm in names:
                Fp = winsor_cs(PANELS[fname][nm])
                ev = evaluate(Fp, FWD, rd)
                if ev is None:
                    rows.append([cat, nm] + [np.nan] * 5)
                    continue
                s = summarize(ev[0], ev[1], ev[2], P['ppy'])
                details[nm] = (ev[2], s)
                rows.append([cat, nm, s['IR'], s['RankIR'], s['IC'], s['RankIC'], s['LS']])
        tab = pd.DataFrame(rows, columns=['因子类型', '因子名称', 'IR', 'Rank IR', 'IC', 'Rank IC', '多空收益'])
        tab['有效'] = np.where((tab['IR'].abs() >= IR_THRESHOLD) & (tab['Rank IR'].abs() >= IR_THRESHOLD), '★', '')
        show = tab.copy()
        for c in ['IR', 'Rank IR', 'IC', 'Rank IC']:
            show[c] = show[c].round(2)
        show['多空收益'] = (show['多空收益'] * 100).round(1).astype(str) + '%'
        print('表: %s因子的有效性测试结果 (多空收益为年化, IC为负的因子已反向)' % fname)
        print(show.to_string(index=False))

        eff = tab[tab['有效'] == '★']
        print('\n%s有效因子(|IR|>=%.2f 且 |RankIR|>=%.2f): %s' % (
            fname, IR_THRESHOLD, IR_THRESHOLD, '、'.join(eff['因子名称']) if len(eff) else '无'))
        print('各类因子 |IR| 均值:')
        print(tab.groupby('因子类型', sort=False)['IR'].apply(lambda x: x.abs().mean()).round(3).to_string())

        # 五组分层净值
        sel = tab.reindex(tab['IR'].abs().sort_values(ascending=False).index).head(PLOT_TOP)['因子名称'].tolist()
        nav_rows = []
        for nm in sel:
            G = details[nm][0]
            nav = (1 + G.fillna(0)).cumprod()
            nav_rows.append([nm] + list(nav.iloc[-1].round(3).values))
        print('\n五组分层期末净值 (|IR|最大的%d个因子; group1=因子值最高组, group5=最低组):' % PLOT_TOP)
        print(pd.DataFrame(nav_rows, columns=['因子'] + ['group%d' % (i + 1) for i in range(N_GROUP)]).to_string(index=False))

        if plt is not None and sel:
            ncol = 2
            nrow = int(np.ceil(len(sel) / float(ncol)))
            fig, axes = plt.subplots(nrow, ncol, figsize=(14, 3.6 * nrow))
            axes = np.array(axes).reshape(-1)
            for ax, nm in zip(axes, sel):
                G = details[nm][0]
                nav = (1 + G.fillna(0)).cumprod()
                nav.loc[G.index[0] - pd.Timedelta(days=1)] = 1.0
                nav = nav.sort_index()
                nav.plot(ax=ax, linewidth=1.3)
                ax.set_title('%s %s五组分层' % (fname, nm))
                ax.grid(alpha=0.3)
            for ax in axes[len(sel):]:
                ax.axis('off')
            plt.tight_layout()
            plt.show()
        all_res[fname] = tab

    # 7) 汇总结论
    print('\n' + '=' * 100)
    print('【汇总】月频 vs 周频 (IR / 多空年化), 便于与研报结论对照 (期限因子稳健; 库存因子周频更强; 动量周频失效; 持仓因子无效)')
    m = all_res[list(all_res.keys())[0]][['因子类型', '因子名称', 'IR', '多空收益']]
    for fname in list(all_res.keys())[1:]:
        w = all_res[fname][['因子名称', 'IR', '多空收益']]
        m = m.merge(w, on='因子名称', suffixes=('_' + list(all_res.keys())[0], '_' + fname))
    for c in m.columns:
        if c.startswith('IR'):
            m[c] = m[c].round(2)
        if c.startswith('多空收益'):
            m[c] = (m[c] * 100).round(1).astype(str) + '%'
    print(m.to_string(index=False))


main()
```
